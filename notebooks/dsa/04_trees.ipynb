# Trees

Trees are hierarchial data structures. It means, in contrast of the previous data
structures we studied, trees store data in a non-sequential manner. They consists of
nodes connected with edges, forming a hierarchial structure with a root node at the top
and leafs at the bottom.

## Terminology

To understand trees, we need to first understand the basic concepts and set up our
terminology. Here are list of terms associated with a tree:

- **Node**: Any data structure that actually stores the data. Each element in the tree.
- **Edge**: The connections between two nodes in the tree.
- **Root**: The top most node in the tree. The starting point of the tree.
- **Parent**: A node that has one or more child nodes.
- **Child**: Nodes directly connected to a parent node.
- **Leaf**: Node without any children. Located at the bottom of the tree.
- **Degree**: The total number of children of a given node.
- **Level**: The root is level 0, its children are in level 1, and so on.
- **Height**: Total numbers of the tree nodes, in the longest path of the tree. But,
  this is my convention, some people prefer the total number of edges.
- **Depth**: Total numbers of the tree edges, from the root of the tree to that node.
- **Diameter**: The diameter of a tree is defined as the number of edges in the longest
  path between two any nodes.
- **Size**: The size of a tree, is the total number of nodes present in the tree.

## Types of Trees

There are several types of trees, each suitable for a group of specific tasks.

- Binary Tree: A tree which every node of it has at most two children, the right child
  and the left child.
- Binary Search Tree(BST): A binary tree which the left child node contains only values
  less than the node's value, and the right child contains only values greater than
  (or equal to) the node's value. In a more technical term, a binary search tree has
  two specific ordering properties:
    - Left subtree property: All nodes in the left subtree of any given node contains
      values lower than the node's value.
    - Right subtree property: All nodes in the right subtree of any given node contains
      values greater than (or equal to) the node's value.
- Balanced Tree: A tree where the heights of the left and right subtrees of any node
  differ by at most a small, bounded amount(usually 1). We have AVL and red-black trees
  here.
- Heap: A tree ds where the parent node is always greater than or less than its child
  nodes, depending the heap being max heap or min heap.
- Trie(Prefix Tree): A tree used for efficient retrieval of a key in large set of
  strings, where each node represents a common prefix.

## Tree Traversal

The method for visiting all nodes of a tree is called `tree traversal`. Unlike the
linear data structures(such as linked-lists or arrays), the trees and non-linear data
structures can not be traversed with one go, so we should use other algorithms and
techniques. There are two main ways of traversal in trees world. 

First, let's look at the Node implementation of a binary tree:

```python
class Node[T]:
    """Represent a Node in a binary tree structure."""

    def __init__(self, data: T) -> None:
        """Instantiate a node with the given data."""
        self.left_child: Node[T] | None
        self.data = data
        self.right_child: Node[T] | None

    def __str__(self) -> str:
        """Return the string representation of the class."""
        return f"TreeNode <{self.data}>"

    def __repr__(self) -> str:
        """Return the official string representation of the class."""
        return f"Node({self.data})"
```

### Depth-First Search(DFS)

In depth-first traversal, we start at the root node and explore nodes far down to the
one branch before backtracking. We use a recursive approach, or we can use a stack to
keep track of the nodes explicitly. There are three forms of depth-first traversal:

- **pre-order**: In the pre-order traversal, we do these steps:

  1. We visit the root node.
  1. Then we visit all nodes of the lef subtree.
  3. At last, we visit the right subtree.

- **in-order**: In the in-order traversal, we do these steps:

  1. We first visit all nodes of the lef subtree.
  2. Then we visit the root node.
  3. Last but not least, we visit the right subtree.

- **post-order**: In the pre-order traversal, we do these steps:

  1. We first visit all nodes of the lef subtree.
  3. Then, we visit the right subtree.
  2. In the end, we visit the root node.

The main characteristic of the depth-first traversal is going as deep as possible in
each branch before backtracking.

**Time and space complexities**: Use less memory than BFS because it doesn't need to
store every node at each level. For balanced trees, the space complexity is O(h), where
h is the height of the tree.

Let's look at the implementations. There are codes and examples that will make you learn
all the things we said.

```python
from collections.abc import Generator

class Node[T]:
    """Represent a Node in a binary tree structure."""

    def __init__(self, data: T) -> None:
        """Instantiate a node with the given data."""
        self.left_child: Node[T] | None = None
        self.data = data
        self.right_child: Node[T] | None = None

    def __str__(self) -> str:
        """Return the string representation of the class."""
        return f"TreeNode <{self.data}>"

    def __repr__(self) -> str:
        """Return the official string representation of the class."""
        return f"Node({self.data})"


def pre_order_traversal[T](root_node: Node[T] | None) -> Generator[T]:
    """
    Perform a pre-order traversal on the binary tree rooted at 'root_node'.

    Parameters
    ----------
    root_node : TreeNode
        The root node of the binary tree to traverse.
    """
    if not root_node:
        return
    yield root_node.data
    yield from pre_order_traversal(root_node=root_node.left_child)
    yield from pre_order_traversal(root_node=root_node.right_child)


def in_order_traversal[T](root_node: Node[T] | None) -> Generator[T]:
    """
    Perform an in-order traversal on the binary tree rooted at root_node.

    Parameters
    ----------
    root_node : TreeNode
        The root node of the binary tree to traverse.
    """
    if not root_node:
        return
    yield from pre_order_traversal(root_node=root_node.left_child)
    yield root_node.data
    yield from pre_order_traversal(root_node=root_node.right_child)


def post_order_traversal[T](root_node: Node[T] | None) -> Generator[T]:
    """
    Perform a post-order traversal on the binary tree rooted at 'root_node'.

    Parameters
    ----------
    root_node : TreeNode
        The root node of the binary tree to traverse.
    """
    if not root_node:
        return
    yield from pre_order_traversal(root_node=root_node.left_child)
    yield from pre_order_traversal(root_node=root_node.right_child)
    yield root_node.data


if __name__ == "__main__":
    A = Node("A")
    B = Node("B")
    C = Node("C")
    D = Node("D")
    E = Node("E")
    F = Node("F")
    G = Node("G")
    H = Node("H")

    A.left_child = B
    A.right_child = C
    B.left_child = D
    B.right_child = E
    D.left_child = G
    D.right_child = H
    A.right_child = C
    C.right_child = F

    # This is what the above tree would look like:
    #        A
    #       / \
    #      B  C
    #     / \  \
    #    D  E  F
    #   / \
    #  G  H

    print("pre-order traversal:")
    for data in pre_order_traversal(A):
        print(data)  # A-B-D-G-H-E-C-F

    print("in-order traversal:")
    for data in in_order_traversal(A):
        print(data)  # G-D-H-B-E-A-C-F

    print("post-order traversal:")
    for data in post_order_traversal(A):
        print(data)  # G-H-D-E-B-F-C-A
```

Easy, right? You can see a great usage of `yield from ` and `generators` in Python in
the above example. Let's see what is the other way to traverse a tree and how we could
implement it.

### Breadth-First Search(BFS)

Breadth-first search, traverses the tree level by level. It starts from the root node
and then explores all the nodes at the present depth level before moving on to the nodes
at the next depth level. It is also called `level-order traversal` because it processes
nodes level by level.

**Time and space complexity**: It needs more space than dfs because it stores all the
nodes at each level. For a complete binary tree, the the space complexity can go up to
O(n) where the n is the number of nodes.

Let's so how we could implement it:

```python
from collections import deque
from collections.abc import Generator


class Node[T]:
    """Represent a Node in a binary tree structure."""

    def __init__(self, data: T) -> None:
        """Instantiate a node with the given data."""
        self.left_child: Node[T] | None = None
        self.data = data
        self.right_child: Node[T] | None = None

    def __str__(self) -> str:
        """Return the string representation of the class."""
        return f"TreeNode <{self.data}>"

    def __repr__(self) -> str:
        """Return the official string representation of the class."""
        return f"Node({self.data})"


def breadth_first_traversal[T](root_node: Node[T] | None) -> Generator[T]:
    """
    Perform a breadth-first traversal on the binary tree rooted at 'root_node'.

    Parameters
    ----------
    root_node : TreeNode
        The root node of the binary tree to traverse.

    Yields
    ------
    T
        The data of each node in breadth-first order.
    """
    if root_node is None:
        return None

    queue = deque()
    queue.append(root_node)
    while queue:
        current_node: Node[T] = queue.popleft()
        yield current_node.data
        if current_node.left_child:
            queue.append(current_node.left_child)
        if current_node.right_child:
            queue.append(current_node.right_child)

if __name__ == "__main__":
    A = Node("A")
    B = Node("B")
    C = Node("C")
    D = Node("D")
    E = Node("E")
    F = Node("F")
    G = Node("G")
    H = Node("H")

    A.left_child = B
    A.right_child = C
    B.left_child = D
    B.right_child = E
    D.left_child = G
    D.right_child = H
    A.right_child = C
    C.right_child = F

    # This is what the above tree would look like:
    #        A
    #       / \
    #      B  C
    #     / \  \
    #    D  E  F
    #   / \
    #  G  H

    print("breadth-first traversal:")
    for data in breadth_first_traversal(A):
        print(data)
```

## Binary Tree

Collection of nodes, where the nodes in the tree can only have zero, one, or two child
nodes.

- **Balanced binary tree**: For every node in the tree, the heights of its left and
  right subtrees differ by at most a specified small number(usually 1).
- **Full binary tree**: When each node of a binary three has zero or two children,
  and there is no node with one child. **Full binary tree focuses on child counts (0 or
  two).**
- **Complete binary tree**: It is a from of binary tree where the tree is completely filled,
  except the bottom level, which should be filled from left to right. **So, this type
  of binary trees focuses on filling levels strictly left to right.**
- **Perfect Binary Tree**: A perfect binary tree is a binary tree with every node has two
  children, except the leaf nodes, which have zero children.

> **Perfect** implies **full**, but not vice-versa.

### The Core Skill: Designing Recursive Return Values

When I write a recursive function on a tree, the recursion naturally goes "down" to the
leaves and then the results bubble "up" through the call stack. So an important question
appears here:

**At each node, what information does this call need to pass back to its parent, so the
parent can do the job?**

This question is not the same as **What's the final answer to the problem?**. The
final answer to the problem is often a side effect of computing right **handback**
value at every node, but the handback value itself might be something the original
problem statement never mentions.

#### Computing Diameter and Height of the Binary Tree

So, let's ask ourselves the right handback question when computing the diameter:

**At any given node, the only thing a parent needs from a child to compute things
involving that child is: How tall is the child's subtree?**

The answer is simply height. Right height and left height of every node should be
calculated and summed up, and at the end, the maximum sum is the diameter of the tree.
A fun thing here, is that *height* is the side effect of our calculation, so we can
calculate diameter and height at one go:

- **Handback value**: The subtree's height.
- **Side effect at each node**: Is `left_height + right_height` bigger than the best
  diameter I've seen so far?

The **pattern** most of the time is: `One quantity gets returned up in the call chain
(because the parent structurally needs it), and a different quantity gets tracked as a
runner-up/accumulator (because it doesn't need to flow through the recursion's return
channel at all, it just need to be check at every node).

Another **pattern** and **questions** worth mentioning, are:
- **What does the base case (None) should return?**
- **What does a leaf returns?** It is often: **What does the general case returns when
  both children are `None`?**.
- **What does an internal node compute from its children's returned value, and what does
  it return upward?**
- **Is there any second piece of information that doesn't need tracking, which doesn't
  belong in the return value?** This is the **side accumulator** case, like the time
  the maximum of values are needed.

Let's look at these concepts and abstract explanations, concretely:

```python
from typing import Annotated, Self

type PositiveInt = Annotated[int, "Value must be >= 0"]
type Height = Annotated[PositiveInt, "The height of the tree"]
type Diameter = Annotated[PositiveInt, "The diameter of the tree"]


class TreeNode:
    def __init__(
            self,
            val: int = 0,
            left: Self | None = None,
            right: Self | None = None,
        ) -> None:
        """Initialize a tree node."""
        self.val = val
        self.left = left
        self.right = right

    def __str__(self) -> str:
        """Return the string representation of the class."""
        return f"TreeNode <{self.val}>"

    def __repr__(self) -> str:
        """Return the official string representation of the class."""
        return f"TreeNode({self.val})"


class TreeService:
    @staticmethod
    def get_size(root: TreeNode | None) -> PositiveInt:
        if root is None:
            return 0
        return 1 + TreeService.get_size(root.left) + TreeService.get_size(root.right)

    @staticmethod
    def get_depth(root: TreeNode | None, target: TreeNode) -> PositiveInt | None:
        if root is None:
            return None

        if root is target:
            return 0

        left_depth = TreeService.get_depth(root=root.left, target=target)
        if left_depth is not None:
            return 1 + left_depth
        right_depth = TreeService.get_depth(root=root.right, target=target)
        if right_depth is not None:
            return 1 + right_depth

        return None

    @staticmethod
    def get_height_and_diameter(root: TreeNode | None) -> tuple[Height, Diameter]:
        if not root:  # Base case
            return 0, 0

        left_height, left_diameter = TreeService.get_height_and_diameter(root.left)
        right_height, right_diameter = TreeService.get_height_and_diameter(root.right)

        # Height is conventionally set to the total number of nodes from root to the
        # deepest node in the tree.
        height = 1 + max(left_height, right_height)

        diameter_through_root = left_height + right_height
        diameter = max(diameter_through_root, left_diameter, right_diameter)

        return height, diameter

    @staticmethod
    def is_balanced(root: TreeNode | None) -> bool:
        pass

    @staticmethod
    def is_full(root: TreeNode | None) -> bool:
        pass

    @staticmethod
    def is_complete(root: TreeNode | None) -> bool:
        pass

    @staticmethod
    def is_perfect(root: TreeNode | None) -> bool:
        pass


if __name__ == "__main__":
    A = TreeNode("A")
    B = TreeNode("B")
    C = TreeNode("C")
    D = TreeNode("D")
    E = TreeNode("E")
    F = TreeNode("F")
    G = TreeNode("G")
    H = TreeNode("H")

    A.left = B
    A.right = C
    B.left = D
    B.right = E
    D.left = G
    D.right = H
    A.right = C
    C.right = F

    # This is what the above tree would look like:
    #        A
    #       / \
    #      B  C
    #     / \  \
    #    D  E  F
    #   / \
    #  G  H

print(TreeService.get_size(root=A))
print(TreeService.get_depth(root=A, target=H))
print(TreeService.get_height_and_diameter(root=A))
```
