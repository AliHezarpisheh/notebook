# Trees

Trees are hierarchial data structures. It means, in contrast of the previous data
structures we studied, trees store data in a non-sequential manner. They consists of
nodes connected with edges, forming a hierarchial structure with a root node at the top
and leafs at the bottom.

## Terminology

To understand trees, we need to first understand the basic concepts and set up our
terminology. Here are list of terms associated with a tree:

- **Node**: Any data structure that actually stores the data. Each element in the tree.
- **Edge**: The connections between two nodes in the tree.
- **Root**: The top most node in the tree. The starting point of the tree.
- **Parent**: A node that has one or more child nodes.
- **Child**: Nodes directly connected to a parent node.
- **Leaf**: Node without any children. Located at the bottom of the tree.
- **Degree**: The total number of children of a given node.
- **Level**: The root is level 0, its children are in level 1, and so on.
- **Height**: Total numbers of the tree nodes, in the longest path of the tree. But,
  this is my convention, some people prefer the total number of edges.
- **Depth**: Total numbers of the tree edges, from the root of the tree to that node.
- **Diameter**: The diameter of a tree is defined as the number of edges in the longest
  path between two any nodes.
- **Size**: The size of a tree, is the total number of nodes present in the tree.

> 🔼 The height of a node in a tree refers to the maximum number of edges on the longest
> path from that node down to a leaf. Every leaf has height 0.

> 🔽 The depth of a node in a tree refers to the number of edges from the root to that
> node. The root has depth 0.

## Types of Trees

There are several types of trees, each suitable for a group of specific tasks.

- Binary Tree: A tree which every node of it has at most two children, the right child
  and the left child.
- Binary Search Tree(BST): A binary tree which the left child node contains only values
  less than the node's value, and the right child contains only values greater than
  (or equal to) the node's value. In a more technical term, a binary search tree has
  two specific ordering properties:
    - Left subtree property: All nodes in the left subtree of any given node contains
      values lower than the node's value.
    - Right subtree property: All nodes in the right subtree of any given node contains
      values greater than (or equal to) the node's value.
- Balanced Tree: A tree where the heights of the left and right subtrees of any node
  differ by at most a small, bounded amount(usually 1). We have AVL and red-black trees
  here.
- Heap: A tree ds where the parent node is always greater than or less than its child
  nodes, depending the heap being max heap or min heap.
- Trie(Prefix Tree): A tree used for efficient retrieval of a key in large set of
  strings, where each node represents a common prefix.

## Tree Traversal

The method for visiting all nodes of a tree is called `tree traversal`. Unlike the
linear data structures(such as linked-lists or arrays), the trees and non-linear data
structures can not be traversed with one go, so we should use other algorithms and
techniques. There are two main ways of traversal in trees world. 

First, let's look at the Node implementation of a binary tree:

````python
class TreeNode[T]:
    """Represent a Node in a binary tree structure."""

    def __init__(self, data: T) -> None:
        """Instantiate a node with the given data."""
        self.left_child: TreeNode[T] | None
        self.data = data
        self.right_child: TreeNode[T] | None

    def __str__(self) -> str:
        """Return the string representation of the class."""
        return f"TreeNode <{self.data}>"

    def __repr__(self) -> str:
        """Return the official string representation of the class."""
        return f"Node({self.data})"
````

### Depth-First Search(DFS)

In depth-first traversal, we start at the root node and explore nodes far down to the
one branch before backtracking. We use a recursive approach, or we can use a stack to
keep track of the nodes explicitly. There are three forms of depth-first traversal:

- **pre-order**: In the pre-order traversal, we do these steps:

  1. We visit the root node.
  1. Then we visit all nodes of the lef subtree.
  3. At last, we visit the right subtree.

- **in-order**: In the in-order traversal, we do these steps:

  1. We first visit all nodes of the lef subtree.
  2. Then we visit the root node.
  3. Last but not least, we visit the right subtree.

- **post-order**: In the pre-order traversal, we do these steps:

  1. We first visit all nodes of the lef subtree.
  3. Then, we visit the right subtree.
  2. In the end, we visit the root node.

The main characteristic of the depth-first traversal is going as deep as possible in
each branch before backtracking.

**Time and space complexities**: Use less memory than BFS because it doesn't need to
store every node at each level. For balanced trees, the space complexity is O(h), where
h is the height of the tree.

Let's look at the implementations. There are codes and examples that will make you learn
all the things we said.

````python
from collections.abc import Generator

class TreeNode[T]:
    """Represent a Node in a binary tree structure."""

    def __init__(self, data: T) -> None:
        """Instantiate a node with the given data."""
        self.left_child: TreeNode[T] | None = None
        self.data = data
        self.right_child: TreeNode[T] | None = None

    def __str__(self) -> str:
        """Return the string representation of the class."""
        return f"TreeTreeNode <{self.data}>"

    def __repr__(self) -> str:
        """Return the official string representation of the class."""
        return f"TreeNode({self.data})"


def pre_order_traversal[T](root_node: TreeNode[T] | None) -> Generator[T]:
    """
    Perform a pre-order traversal on the binary tree rooted at 'root_node'.

    Parameters
    ----------
    root_node : TreeNode
        The root node of the binary tree to traverse.
    """
    if not root_node:
        return
    yield root_node.data
    yield from pre_order_traversal(root_node=root_node.left_child)
    yield from pre_order_traversal(root_node=root_node.right_child)


def in_order_traversal[T](root_node: TreeNode[T] | None) -> Generator[T]:
    """
    Perform an in-order traversal on the binary tree rooted at root_node.

    Parameters
    ----------
    root_node : TreeNode
        The root node of the binary tree to traverse.
    """
    if not root_node:
        return
    yield from pre_order_traversal(root_node=root_node.left_child)
    yield root_node.data
    yield from pre_order_traversal(root_node=root_node.right_child)


def post_order_traversal[T](root_node: TreeNode[T] | None) -> Generator[T]:
    """
    Perform a post-order traversal on the binary tree rooted at 'root_node'.

    Parameters
    ----------
    root_node : TreeNode
        The root node of the binary tree to traverse.
    """
    if not root_node:
        return
    yield from pre_order_traversal(root_node=root_node.left_child)
    yield from pre_order_traversal(root_node=root_node.right_child)
    yield root_node.data


if __name__ == "__main__":
    A = TreeNode("A")
    B = TreeNode("B")
    C = TreeNode("C")
    D = TreeNode("D")
    E = TreeNode("E")
    F = TreeNode("F")
    G = TreeNode("G")
    H = TreeNode("H")

    A.left_child = B
    A.right_child = C
    B.left_child = D
    B.right_child = E
    D.left_child = G
    D.right_child = H
    A.right_child = C
    C.right_child = F

    # This is what the above tree would look like:
    #        A
    #       / \
    #      B  C
    #     / \  \
    #    D  E  F
    #   / \
    #  G  H

    print("pre-order traversal:")
    for data in pre_order_traversal(A):
        print(data)  # A-B-D-G-H-E-C-F

    print("in-order traversal:")
    for data in in_order_traversal(A):
        print(data)  # G-D-H-B-E-A-C-F

    print("post-order traversal:")
    for data in post_order_traversal(A):
        print(data)  # G-H-D-E-B-F-C-A
````

Easy, right? You can see a great usage of `yield from ` and `generators` in Python in
the above example. Let's see what is the other way to traverse a tree and how we could
implement it.

### Breadth-First Search(BFS)

Breadth-first search, traverses the tree level by level. It starts from the root node
and then explores all the nodes at the present depth level before moving on to the nodes
at the next depth level. It is also called `level-order traversal` because it processes
nodes level by level.

**Time and space complexity**: It needs more space than dfs because it stores all the
nodes at each level. For a complete binary tree, the the space complexity can go up to
O(n) where the n is the number of nodes.

Let's so how we could implement it:

````python
from collections import deque
from collections.abc import Generator


class Tree[T]:
    """Represent a Node in a binary tree structure."""

    def __init__(self, data: T) -> None:
        """Instantiate a node with the given data."""
        self.left_child: Tree[T] | None = None
        self.data = data
        self.right_child: Tree[T] | None = None

    def __str__(self) -> str:
        """Return the string representation of the class."""
        return f"TreeNode <{self.data}>"

    def __repr__(self) -> str:
        """Return the official string representation of the class."""
        return f"Tree({self.data})"


def breadth_first_traversal[T](root_node: Tree[T] | None) -> Generator[T]:
    """
    Perform a breadth-first traversal on the binary tree rooted at 'root_node'.

    Parameters
    ----------
    root_node : TreeNode
        The root node of the binary tree to traverse.

    Yields
    ------
    T
        The data of each node in breadth-first order.
    """
    if root_node is None:
        return None

    queue = deque()
    queue.append(root_node)
    while queue:
        current_node: Tree[T] = queue.popleft()
        yield current_node.data
        if current_node.left_child:
            queue.append(current_node.left_child)
        if current_node.right_child:
            queue.append(current_node.right_child)

if __name__ == "__main__":
    A = Tree("A")
    B = Tree("B")
    C = Tree("C")
    D = Tree("D")
    E = Tree("E")
    F = Tree("F")
    G = Tree("G")
    H = Tree("H")

    A.left_child = B
    A.right_child = C
    B.left_child = D
    B.right_child = E
    D.left_child = G
    D.right_child = H
    A.right_child = C
    C.right_child = F

    # This is what the above tree would look like:
    #        A
    #       / \
    #      B  C
    #     / \  \
    #    D  E  F
    #   / \
    #  G  H

    print("breadth-first traversal:")
    for data in breadth_first_traversal(A):
        print(data)
````

## General Tree (n-ary Tree)

Generic trees are a collection of nodes where each node is a data structure consists of
records and a list of references to its children. Unlike linked lists, each node stores
the address of multiple nodes. 

![n-ary tree](./images/generic-tree.png)

We have three approaches for creating the tree node here that I have covered below.

### One New Technique - Left-Child Right-Siblings(LCRS)

Here, `left` points to the first child, and `right` points to the next siblings. A trick
useful when you want to reuse binary tree mechanisms without writing separate code.

### Use Cases

File systems, DOM/XML trees, org charts, category hierarchies, JSON schemas, and Tries.

````python
class NaryNode[T]:
    """Represent a node in a general tree, using a simple approach."""

    def __init__(self, value: T):
        """Instantiate a node with the given value."""
        self.value = value
        self.first_child: NaryNode[T] | None = None
        self.second_child: NaryNode[T] | None = None
        self.third_child: NaryNode[T] | None = None
        self.fourth_child: NaryNode[T] | None = None
        self.fifth_child: NaryNode[T] | None = None


class NaryNode[T]:
    """Represent a Node in general tree, using a dynamic array for storing children."""

    def __init__(self, value: T) -> None:
        """Instantiate a node with the given value."""
        self.value = value
        self.children: list[NaryNode[T]] | None


class NaryNode[T]:
    """Represent a Node in a general tree, using first child/next sibling approach."""

    def __init__(self, value: T) -> None:
        """Instantiate a node with the given value."""
        self.value = value
        self.first_child: TreeNode | None = None
        self.next_sibling: NaryNode[T] | None = None
````

````python
from __future__ import annotations

from collections import deque
from dataclasses import dataclass, field


@dataclass
class NaryNode[T]:
    """Represent a node in a general tree."""

    value: T
    children: list[NaryNode] = field(default_factory=list)

    def add_child(self, child: NaryNode) -> NaryNode:
        self.children.append(child)
        return child

    def __repr__(self) -> str:
        return f"NaryNode({self.value!r})"


class NaryTreeService[T]:
    @staticmethod
    def traverse_preorder(root: NaryNode | None) -> list[T]:
        """Root, then each child subtree left to right. O(n) time, O(h) space."""
        if root is None:
            return []
        result = [root.value]
        for child in root.children:
            result.extend(NaryTreeService.traverse_preorder(root=child))
        return result

    @staticmethod
    def traverse_preorder_iterative(root: NaryNode | None) -> list[T]:
        """Same order as `preorder`, without recursion (avoids stack-depth limits)."""
        if root is None:
            return []
        result: list[T] = []
        stack: list[NaryNode[T]] = [root]
        while stack:
            node = stack.pop()
            result.append(node.value)
            stack.extend(reversed(node.children))
        return result

    @staticmethod
    def traverse_postorder(root: NaryNode | None) -> list[T]:
        """Each child subtree left to right, then root."""
        if root is None:
            return []
        result: list[T] = []
        for child in root.children:
            result.extend(NaryTreeService.traverse_postorder(root=child))
        result.append(root.value)
        return result

    @staticmethod
    def traverse_postorder_iterative(root: NaryNode | None) -> list[T]:
        """
        Push root, record in a 'reverse-ish' order, then flip (mirror of the
        preorder trick: root-last instead of root-first).
        """
        if root is None:
            return []
        result: list[T] = []
        stack: list[NaryNode[T]] = [root]
        while stack:
            node = stack.pop()
            result.append(node.value)
            stack.extend(node.children)
        return result[::-1]

    @staticmethod
    def traverse_level_order(root: NaryNode | None) -> list[T]:
        """BFS, grouped by depth. O(n) time, O(w) space where w = max tree width."""
        if root is None:
            return []

        result: list[list[T]] = []
        queue: deque = deque([root])
        while queue:
            level = []
            for _ in range(len(queue)):
                node = queue.popleft()
                level.append(node.value)
                queue.extend(node.children)
            result.append(level)
        return result

    @staticmethod
    def get_height(root: NaryNode | None) -> int:
        """Number of nodes on the longest root-to-leaf path. Empty tree -> 0."""
        if root is None:
            return 0
        if not root.children:
            return 1
        return 1 + max(NaryTreeService.get_height(child) for child in root.children)

    @staticmethod
    def get_nodes_count(root: NaryNode | None) -> int:
        """Get counts of all nodes in the tree."""
        if root is None:
            return 0
        return 1 + sum(
            NaryTreeService.get_nodes_count(child) for child in root.children
        )

    @staticmethod
    def find(root: NaryNode | None, target: T) -> NaryNode | None:
        """DFS lookup by value. Returns the first match or None."""
        if root is None:
            return None
        if root is target:
            return root
        for child in root.children:
            hit = NaryTreeService.find(root=child, target=target)
            if hit is not None:
                return hit
        return None

    @staticmethod
    def serialize(root: NaryNode | None) -> str:
        """Serialize tree into a sequence of flat tokens."""
        tokens: list[str] = []

        def dfs(node: NaryNode | None) -> None:
            if node is None:
                return None
            tokens.append(str(node.value))
            tokens.append(str(len(node.children)))
            for child in node.children:
                dfs(node=child)

        dfs(node=root)
        return " ".join(tokens)

    @staticmethod
    def deserialize(data: str) -> NaryNode | None:
        """Deserialize data into a n-ary tree."""
        if not data:
            return None
        tokens = iter(data.split())

        def dfs() -> NaryNode:
            value = next(tokens)
            children_count = int(next(tokens))
            node = NaryNode(value=value)
            for _ in range(children_count):
                node.add_child(dfs())
            return node

        return dfs()


if __name__ == "__main__":
    #        A
    #      / | \
    #     B  C  D
    #    / \      \
    #   E   F      G
    root: NaryNode[str] = NaryNode("A")
    b = root.add_child(NaryNode("B"))
    c = root.add_child(NaryNode("C"))
    d = root.add_child(NaryNode("D"))
    b.add_child(NaryNode("E"))
    b.add_child(NaryNode("F"))
    d.add_child(NaryNode("G"))

    service = NaryTreeService()
    print(service.traverse_preorder(root=root))
    print(service.traverse_postorder(root=root))
    print(service.traverse_preorder_iterative(root=root))
    print(service.traverse_postorder_iterative(root=root))
    print(service.traverse_level_order(root=root))

    print(service.get_height(root=root))
    print(service.get_nodes_count(root=root))
    print(service.find(root=root, target=d))

    serialized = service.serialize(root=root)
    print(serialized)
    print(service.deserialize(data=serialized))
````

````python
from __future__ import annotations

from dataclasses import dataclass, field


@dataclass
class NaryNode[T]:
    """Represent a node in a general tree."""

    value: T
    children: list[NaryNode] = field(default_factory=list)

    def add_child(self, child: NaryNode) -> NaryNode:
        self.children.append(child)
        return child

    def __repr__(self) -> str:
        return f"NaryNode({self.value!r})"


@dataclass
class LCRSNode[T]:
    """Represent a node in a general tree, with left-child right-sibling approach."""

    value: T
    left_child: LCRSNode[T] | None = None
    next_sibling: LCRSNode[T] | None = None

    def __repr__(self) -> str:
        return f"LCRSNode({self.value!r})"


def convert_nary_to_lcrs[T](root: NaryNode[T] | None) -> LCRSNode[T] | None:
    """Convert a nary tree to a lcrs one."""
    if root is None:
        return None

    lcrs_root = LCRSNode(value=root.value)
    if root.children:
        lcrs_root.left_child = convert_nary_to_lcrs(root=root.children[0])
        current = lcrs_root.left_child
        for sibling in root.children[1:]:
            current.next_sibling = convert_nary_to_lcrs(root=sibling)
            current = current.next_sibling
    return lcrs_root


def convert_lcrs_to_nary[str](root: LCRSNode[str] | None) -> NaryNode[str] | None:
    """Convert a lcrs tree to a nary one."""
    if root is None:
        return None
    nary_node = NaryNode(value=root.value)
    child = root.left_child
    while child is not None:
        nary_node.add_child(child=convert_lcrs_to_nary(child))
        child = child.next_sibling
    return nary_node


if __name__ == "__main__":
    #        A
    #      / | \
    #     B  C  D
    #    / \      \
    #   E   F      G
    root: NaryNode[str] = NaryNode("A")
    b = root.add_child(NaryNode("B"))
    c = root.add_child(NaryNode("C"))
    d = root.add_child(NaryNode("D"))
    b.add_child(NaryNode("E"))
    b.add_child(NaryNode("F"))
    d.add_child(NaryNode("G"))

    print("======== LCRS ========")
    lcrs_root = convert_nary_to_lcrs(root=root)

    print(lcrs_root)
    print(lcrs_root.left_child)
    print(lcrs_root.left_child.next_sibling)
    print(lcrs_root.left_child.next_sibling.next_sibling)
    print(lcrs_root.left_child.left_child)
    print(lcrs_root.left_child.left_child.next_sibling)
    print(lcrs_root.left_child.next_sibling.next_sibling.left_child)

    print("======== Nary ========")
    nary_node = convert_lcrs_to_nary(root=lcrs_root)
    print(nary_node)
    print(nary_node.children)
    print(nary_node.children[0].children)
    print(nary_node.children[1].children)
    print(nary_node.children[2].children)
````

## Binary Tree

Collection of nodes, where the nodes in the tree can only have zero, one, or two child
nodes.

- **Balanced binary tree**: For every node in the tree, the heights of its left and
  right subtrees differ by at most a specified small number(usually 1).
- **Full binary tree**: When each node of a binary three has zero or two children,
  and there is no node with one child. **Full binary tree focuses on child counts (0 or
  two).**
- **Complete binary tree**: It is a from of binary tree where the tree is completely filled,
  except the bottom level, which should be filled from left to right. **So, this type
  of binary trees focuses on filling levels strictly left to right.**
- **Perfect Binary Tree**: A perfect binary tree is a binary tree with every node has two
  children, except the leaf nodes, which have zero children.

> **Perfect** implies **full**, but not vice-versa.

This is where the recursive and traversing thinking in binary trees(not just BSTs) is
created.

### The Core Skill: Designing Recursive Return Values

When I write a recursive function on a tree, the recursion naturally goes "down" to the
leaves and then the results bubble "up" through the call stack. So an important question
appears here:

**At each node, what information does this call need to pass back to its parent, so the
parent can do the job?**

This question is not the same as **What's the final answer to the problem?**. The
final answer to the problem is often a side effect of computing right **handback**
value at every node, but the handback value itself might be something the original
problem statement never mentions.

#### Computing Diameter and Height of the Binary Tree

So, let's ask ourselves the right handback question when computing the diameter:

**At any given node, the only thing a parent needs from a child to compute things
involving that child is: How tall is the child's subtree?**

The answer is simply height. Right height and left height of every node should be
calculated and summed up, and at the end, the maximum sum is the diameter of the tree.
A fun thing here, is that *height* is the side effect of our calculation, so we can
calculate diameter and height at one go:

- **Handback value**: The subtree's height.
- **Side effect at each node**: Is `left_height + right_height` bigger than the best
  diameter I've seen so far?

The **pattern** most of the time is: One quantity gets returned up in the call chain
(because the parent structurally needs it), and a different quantity gets tracked as a
runner-up/accumulator (because it doesn't need to flow through the recursion's return
channel at all, it just need to be check at every node).

Another **pattern** and **questions** worth mentioning, are:
- **What does the base case (None) should return?**
- **What does a leaf returns?** It is often: **What does the general case returns when
  both children are `None`?**.
- **What does an internal node compute from its children's returned value, and what does
  it return upward?**
- **Is there any second piece of information that needs tracking, which doesn't
  belong in the return value?** This is the **side accumulator** case, like the time
  the maximum of values are needed.

Let's look at these concepts and abstract explanations, concretely:

````python
from collections import deque
from typing import Annotated, Self

type PositiveInt = Annotated[int, "Value must be >= 0"]
type Height = Annotated[PositiveInt, "The height of the tree"]
type Diameter = Annotated[PositiveInt, "The diameter of the tree"]


class TreeNode:
    """Represent a node in a binary tree."""

    def __init__(
        self,
        val: int | str = 0,
        left: Self | None = None,
        right: Self | None = None,
    ) -> None:
        """Initialize a tree node."""
        self.val = val
        self.left = left
        self.right = right

    def __str__(self) -> str:
        """Return the string representation of the class."""
        return f"TreeNode <{self.val}>"

    def __repr__(self) -> str:
        """Return the official string representation of the class."""
        return f"TreeNode({self.val})"


class TreeService:
    """Encapsulate various binary trees operations."""

    @staticmethod
    def get_size(root: TreeNode | None) -> PositiveInt:
        """
        Return the total number of nodes in a tree.

        Parameters
        ----------
        root : TreeNode | None
            Root of the tree. ``None`` represents an empty tree.

        Returns
        -------
        PositiveInt
            Number of nodes reachable from ``root``.
        """
        if root is None:
            return 0
        return 1 + TreeService.get_size(root.left) + TreeService.get_size(root.right)

    @staticmethod
    def get_depth(root: TreeNode | None, target: TreeNode) -> PositiveInt | None:
        """
        Return the depth of a node relative to the tree root.

        Parameters
        ----------
        root : TreeNode | None
            Root of the tree to search.
        target : TreeNode
            Node whose depth is requested. The search compares object
            identity rather than node values.

        Returns
        -------
        PositiveInt | None
            Number of edges from ``root`` to ``target``, or ``None`` if the
            target is not contained in the tree.
        """
        if root is None:
            return None

        if root is target:
            return 0

        left_depth = TreeService.get_depth(root=root.left, target=target)
        if left_depth is not None:
            return 1 + left_depth
        right_depth = TreeService.get_depth(root=root.right, target=target)
        if right_depth is not None:
            return 1 + right_depth

        return None

    @staticmethod
    def get_height_and_diameter(root: TreeNode | None) -> tuple[Height, Diameter]:
        """
        Return the height and diameter of a tree.

        Parameters
        ----------
        root : TreeNode | None
            Root of the tree. ``None`` represents an empty tree.

        Returns
        -------
        tuple[Height, Diameter]
            A pair containing the tree height and the maximum number of
            edges on any path between two nodes.

        Notes
        -----
        Height is measured as the number of nodes on the longest path from
        the root to a leaf.
        """
        if root is None:  # Base case
            return 0, 0

        left_height, left_diameter = TreeService.get_height_and_diameter(root.left)
        right_height, right_diameter = TreeService.get_height_and_diameter(root.right)

        # Height is conventionally set to the total number of nodes from root to the
        # deepest node in the tree.
        height = 1 + max(left_height, right_height)

        diameter_through_root = left_height + right_height
        diameter = max(diameter_through_root, left_diameter, right_diameter)

        return height, diameter

    @staticmethod
    def is_balanced(root: TreeNode | None) -> bool:
        """
        Return whether the tree satisfies the height-balanced property.

        Parameters
        ----------
        root : TreeNode | None
            Root of the tree. ``None`` is considered balanced.

        Returns
        -------
        bool
            ``True`` if every node has left and right subtrees whose heights
            differ by at most one; otherwise, ``False``.
        """
        UNBALANCED = -1

        def check(node: TreeNode | None) -> int:
            if node is None:
                return 0

            left_height = check(node.left)
            if left_height == UNBALANCED:
                return UNBALANCED
            right_height = check(node.right)
            if right_height == UNBALANCED:
                return UNBALANCED

            if abs(right_height - left_height) > 1:
                return UNBALANCED

            return 1 + max(left_height, right_height)

        return check(root) != UNBALANCED

    @staticmethod
    def is_full(root: TreeNode | None) -> bool:
        """
        Return whether every node has either zero or two children.

        Parameters
        ----------
        root : TreeNode | None
            Root of the tree. ``None`` is considered full.

        Returns
        -------
        bool
            ``True`` if no node has exactly one child; otherwise,
            ``False``.
        """
        if root is None:
            return True
        if (root.left is None) != (root.right is None):
            return False
        return TreeService.is_full(root.left) and TreeService.is_full(root.right)

    @staticmethod
    def is_complete(root: TreeNode | None) -> bool:
        """
        Return whether the tree is a complete binary tree.

        Parameters
        ----------
        root : TreeNode | None
            Root of the tree. ``None`` is considered complete.

        Returns
        -------
        bool
            ``True`` if every level is completely filled except possibly
            the last, whose nodes occupy the leftmost positions.
        """
        if root is None:
            return True

        total_size = TreeService.get_size(root=root)
        queue: deque[tuple[TreeNode, int]] = deque([(root, 1)])

        while queue:
            node, index = queue.popleft()
            if index > total_size:
                return False
            if node.left is not None:
                queue.append((node.left, index * 2))
            if node.right is not None:
                queue.append((node.right, index * 2 + 1))
        return True

    @staticmethod
    def is_perfect(root: TreeNode | None) -> bool:
        """
        Return whether the tree is a perfect binary tree.

        Parameters
        ----------
        root : TreeNode | None
            Root of the tree. ``None`` is considered perfect.

        Returns
        -------
        bool
            ``True`` if every internal node has exactly two children and
            all leaves are at the same depth; otherwise, ``False``.

        Notes
        -----
        This method is strictly thighted to `get_height_and_diameter` method. If the
        conventions and return values are changed in that method, this function will
        probably break.
        """
        height, _ = TreeService.get_height_and_diameter(root=root)

        def check(node: TreeNode | None, depth: int) -> bool:
            if node is None:
                return True
            if (node.left is None) and (node.right is None):
                return height == depth
            if (node.left is None) or (node.right is None):
                return False
            return check(node=node.left, depth=depth + 1) and check(
                node=node.right, depth=depth + 1
            )

        return check(node=root, depth=1)


if __name__ == "__main__":
    A = TreeNode("A")
    B = TreeNode("B")
    C = TreeNode("C")
    D = TreeNode("D")
    E = TreeNode("E")
    F = TreeNode("F")
    G = TreeNode("G")
    H = TreeNode("H")

    A.left = B
    A.right = C
    B.left = D
    B.right = E
    D.left = G
    D.right = H
    A.right = C
    C.right = F

    # This is what the above tree would look like:
    #        A
    #       / \
    #      B  C
    #     / \  \
    #    D  E  F
    #   / \
    #  G  H

    print(TreeService.get_size(root=A))
    print(TreeService.get_depth(root=A, target=H))
    print(TreeService.get_height_and_diameter(root=A))
    print(TreeService.is_balanced(root=A))
    print(TreeService.is_full(root=A))
    print(TreeService.is_complete(root=A))
    print(TreeService.is_perfect(root=A))
````

### Path Problems

A path in a binary tree is a sequence of nodes where each consecutive pair is connected
by an edge, and no node appears twice. What varies in problems are:

- **Where the path start and end?** Root only, any node, must end at leaf?
- **What are we computing?** Enumerate all paths? Check if a sum exists? Maximize a sum?

#### Root-to-leaf Paths

A leaf is a node without any children. A root-to-leaf path starts exactly at the root and
ends exactly at a leaf.

```txt
        1
       / \
      2   3
     / \
    4   5
```

Root-to-leaf paths: [[1, 2, 4], [1, 2, 5], [1, 3]]

#### Path Sum Variants

Same root-to-leaf constraint, but now we care about sum of values along the path, not
the path structure itself.

The first variant of this problem asks: **Does any root-to-leaf path sum to `target`?**
The second variant wants: **All the root-to-leaf paths that sums to `target`**. The most
difficult variant in terms of implementation, wants: **All downward paths (not 
necessarily root-to-leaf) that sums to `target`.**

#### Maximum Path Sum Between Any Two Nodes

The path can start and end at any nodes! This is where the the `recursive return value +
side accumulator` comes to play.

```txt
        -10
        /  \
       9   20
          /  \
         15   7
```

The max path sum here is: 15 -> 20 -> 7.

Let's implement the methods and functions.

````python
from collections import defaultdict
from typing import Self


class TreeNode:
    """Represent a node in a binary tree."""

    def __init__(
        self,
        val: int | str = 0,
        left: Self | None = None,
        right: Self | None = None,
    ) -> None:
        """Initialize a tree node."""
        self.val = val
        self.left = left
        self.right = right

    def __str__(self) -> str:
        """Return the string representation of the class."""
        return f"TreeNode <{self.val}>"

    def __repr__(self) -> str:
        """Return the official string representation of the class."""
        return f"TreeNode({self.val})"


class PathProblem[T]:
    """Return all root-to-leaf paths as lists of node values."""

    @staticmethod
    def get_root_to_leaf_paths(root: TreeNode | None) -> list[list[T]]:
        """
        Return all root-to-leaf paths as lists of node values.

        Design questions answered:
        - None return  : None is never passed directly — we guard before recursing.
        - Leaf return  : append a COPY of the current path to results.
        - Internal node: append value, recurse both sides, pop on return (backtrack).
        - Accumulator  : `results` list lives outside the recursion and is updated
                         as a side effect when we hit a leaf. The recursive helper
                         itself returns nothing — all output goes into `results`.

        The critical bug to avoid: recording `path` itself (a reference) instead of
        `path.copy()`. By the time all recursion unwinds, the original list is empty
        again — so every "recorded" path would point to the same empty list.

        Time : O(n)       — every node visited exactly once
        Space: O(h)       — recursion stack depth, h = tree height
               + O(n * h) — space for all stored paths in the worst case
        """
        result: list[list[T]] = []

        def backtrack(node: TreeNode | None, path: list[T]) -> None:
            if node is None:
                return

            path.append(node.val)

            if node.left is None and node.right is None:
                result.append(path.copy())
            backtrack(node=node.left, path=path)
            backtrack(node=node.right, path=path)

            path.pop()

        backtrack(node=root, path=[])
        return result

    @staticmethod
    def has_path_sum(root: TreeNode | None, target: int) -> bool:
        """
        Return True if any root-to-leaf path sums to `target`.

        Design questions answered:
        - None return  : False — there is no path through a missing node.
        - Leaf return  : True iff the remaining budget exactly hits zero.
        - Internal node: subtract current value, recurse; True if either side hits.
        - Accumulator  : none — the answer flows cleanly through the return value
                         via short-circuit `or`.

        We avoid carrying a running sum downward; instead we subtract and check
        for zero at the leaf. Either direction works; subtraction is slightly
        cleaner because the base condition is a single `== 0` check.

        Time : O(n) worst case, O(log n) best case (balanced tree, found early)
        Space: O(h) recursion stack
        """
        if root is None:
            return False

        remaining = target - root.val
        if root.left is None and root.right is None:
            return remaining == 0
        return PathProblem.has_path_sum(
            root=root.left, target=remaining
        ) or PathProblem.has_path_sum(root=root.right, target=remaining)

    @staticmethod
    def get_all_sum_paths(root: TreeNode | None, target: int) -> list[list[T]]:
        """
        Return all root-to-leaf paths whose values sum to `target`.

        This is flavor 1 (backtracking) + flavor 2a (sum check) combined.
        The structure is identical to root_to_leaf_paths; we just add a guard
        at the leaf so we only record paths that satisfy the sum condition.

        Design questions answered:
        - None return  : guard before recursing, so None is never visited.
        - Leaf return  : record path copy only if `remaining == 0`.
        - Internal node: same backtracking pattern — append, recurse, pop.
        - Accumulator  : `results` list (side effect), `remaining` passed as param.

        Time : O(n)
        Space: O(h) stack + O(k * h) for k matching paths stored
        """
        result: list[list[T]] = []

        def backtrack(root: TreeNode | None, target: int, path: list[T]) -> None:
            if root is None:
                return None

            path.append(root.val)
            remaining = target - root.val

            if (root.left is None and root.right is None) and (remaining == 0):
                result.append(path.copy())
            backtrack(root=root.left, target=remaining, path=path)
            backtrack(root=root.right, target=remaining, path=path)

            path.pop()

        backtrack(root=root, target=target, path=[])
        return result

    @staticmethod
    def count_paths_with_sum(root: TreeNode | None, target: int) -> int:
        """
        Count all downward paths (any start node, any end node, strictly parent
        → child direction) whose values sum to `target`.

        This is fundamentally different from the root-to-leaf variants. The path
        can start and end at any node, but must go strictly downward (no going up).

        KEY INSIGHT — prefix sums on a tree path:
            At any node, the running sum from the root down to here is `prefix`.
            A sub-path ending at this node sums to `target` iff there exists an
            earlier node on the same root-to-here path whose prefix equals
            `prefix - target`. Exactly like "subarray sum equals k" on an array,
            but on a tree path via DFS.

        We maintain `prefix_counts: {prefix_sum → how_many_times_seen}` for the
        current root-to-node path. After visiting both children we *undo* (decrement)
        the current prefix — this is the backtracking step that keeps the dict
        accurate for sibling subtrees that don't share this path.

        Design questions answered:
        - Return value : the helper returns nothing — count lives outside as a side
                         effect (nonlocal int), incremented at each valid path end.
        - None return  : base case — just return, don't update anything.
        - Internal node: update running sum, check dict, recurse both, then undo.
        - Accumulator  : `count` (nonlocal) + `prefix_counts` dict (with undo).

        The initial `prefix_counts[0] = 1` seeds the dict: it accounts for paths
        that start at the root itself. Without it, a prefix that exactly equals
        `target` at some node would incorrectly find zero matches in the dict.

        Time : O(n) — each node visited once, dict operations O(1)
        Space: O(h) — dict holds at most h entries at any point (one per ancestor)
        """
        count = 0
        prefix_counts: defaultdict[int, int] = defaultdict(int)
        prefix_counts[0] = 1

        def dfs(node: TreeNode | None, remaining_sum: int) -> None:
            nonlocal count

            if node is None:
                return None

            remaining_sum += node.val
            count += prefix_counts[remaining_sum - target]

            prefix_counts[remaining_sum] += 1
            dfs(node=node.left, remaining_sum=remaining_sum)
            dfs(node=node.right, remaining_sum=remaining_sum)

            prefix_counts[remaining_sum] -= 1

        dfs(node=root, remaining_sum=0)
        return count

    @staticmethod
    def get_max_path_sum(root: TreeNode | None) -> int:
        """
        Return the maximum sum of any path in the tree. The path may start and
        end at any node; it does not need to pass through the root or reach a leaf.

        This is the hardest path problem and a direct application of the
        "recursive return value + side accumulator" pattern from bullet 2.

        KEY DESIGN DECISION — two separate quantities:

          Return value  : "best single downward arm from this node" — the maximum
                          sum you can get by starting at this node and extending into
                          AT MOST ONE child. This is what the parent needs; it lets
                          the parent extend its own path through this node.

          Side effect   : `global_max` — updated at every node with the "bent path"
                          value (left_arm + node.val + right_arm). This path bends
                          here and cannot be extended further upward, so it doesn't
                          belong in the return value — it just needs to be tracked.

        The `max(..., 0)` floor on child gains means: if a subtree would reduce the
        sum, don't extend into it at all (treat it as if it doesn't exist). This
        handles trees with all-negative values correctly: the global_max ends up as
        the single least-negative node, NOT 0 (because path_through_node uses
        node.val + 0 + 0 = node.val, which is negative, and global_max starts at
        -inf so it will be set to the best single node found).

        Design questions answered:
        - None return  : 0 — contributes nothing to the parent's arm.
        - Leaf return  : node.val (left_gain = right_gain = max(0,0) = 0).
        - Internal node: return node.val + max(left_gain, right_gain, 0).
        - Accumulator  : global_max, updated via side effect at each node.

        Raises ValueError for empty tree — max path sum is undefined with no nodes.

        Time : O(n)
        Space: O(h) recursion stack
        """
        if root is None:
            raise ValueError("Can not get max path sum of an empty tree.")

        global_max = float("-inf")

        def dfs(node: TreeNode | None) -> int | float:
            nonlocal global_max

            breakpoint()
            if node is None:
                return 0

            left_gain = max(dfs(node=node.left), 0)
            right_gain = max(dfs(node=node.right), 0)

            path_through_node = node.val + left_gain + right_gain
            global_max = max(global_max, path_through_node)

            return node.val + max(left_gain, right_gain)

        dfs(node=root)
        return int(global_max)


if __name__ == "__main__":
    A = TreeNode("A")
    B = TreeNode("B")
    C = TreeNode("C")
    D = TreeNode("D")
    E = TreeNode("E")
    F = TreeNode("F")
    G = TreeNode("G")
    H = TreeNode("H")

    A.left = B
    A.right = C
    B.left = D
    B.right = E
    D.left = G
    D.right = H
    A.right = C
    C.right = F

    # This is what the above tree would look like:
    #        A
    #       / \
    #      B  C
    #     / \  \
    #    D  E  F
    #   / \
    #  G  H

    problem = PathProblem()
    print(problem.get_root_to_leaf_paths(root=A))

    A = TreeNode(1)
    B = TreeNode(2)
    C = TreeNode(3)
    D = TreeNode(4)
    E = TreeNode(11)
    F = TreeNode(6)
    G = TreeNode(3)
    H = TreeNode(8)

    A.left = B
    A.right = C
    B.left = D
    B.right = E
    D.left = G
    D.right = H
    A.right = C
    C.right = F

    # This is what the above tree would look like:
    #        1
    #       / \
    #      2  3
    #     / \  \
    #    4  11  6
    #   / \
    #  3  8
    print(problem.has_path_sum(root=A, target=16))
    print(problem.get_all_sum_paths(root=A, target=8))
    print(problem.count_paths_with_sum(root=A, target=14))
    print(problem.get_max_path_sum(root=A))
````

### Comparisons

All these problems share the same fundamental mechanism: **simultaneously walking two
trees (or two parts of the same tree) and comparing structure + values at each step. The
recursion template is almost identical across all four problems here, What change is
which two nodes you pair up at each recursive call.

The key idea behind solving these problems are not asking: *Is tree X?*, instead, you
should reframe the question as a two-argument question: *Are these two nodes/subtrees
mutually Y?*.

#### Same Tree

**Two trees are the same if they have identical structure and identical values at every
corresponding node.**

The question: **Are p and q the roots of identical trees?**

#### Symmetric Tree

**A single tree is symmetric, if the left sub-tree is a mirror image of the right
sub-tree.**

#### Subtree Check

Is tree `sub` is a subtree of `main`.

#### Mirror

Transform a tree into its own mirror image by swapping every node's left and right children.

```txt
Before:        After:
    1              1
   / \            / \
  2   3          3   2
 / \              / \
4   5            5   4
```


````python
from typing import Self


class TreeNode:
    """Represent a node in a binary tree."""

    def __init__(
        self,
        val: int | str = 0,
        left: Self | None = None,
        right: Self | None = None,
    ) -> None:
        """Initialize a tree node."""
        self.val = val
        self.left = left
        self.right = right

    def __str__(self) -> str:
        """Return the string representation of the class."""
        return f"TreeNode <{self.val}>"

    def __repr__(self) -> str:
        """Return the official string representation of the class."""
        return f"TreeNode({self.val})"


class TreeComparisonProblem[T]:
    """
    Structural comparison operations on plain binary trees.

    The unifying idea across all four methods:
        Reframe every single-tree question into a two-node question —
        "are these two nodes mutually X?" — then walk both simultaneously.
        What changes between methods is only WHICH two nodes get paired
        at each recursive step.
    """

    @staticmethod
    def is_same(p: TreeNode | None, q: TreeNode | None) -> bool:
        """
        Return True if trees rooted at p and q are structurally identical
        with equal values at every corresponding node.

        Design questions:
        - None return   : both None → True (two empty trees are the same).
                          One None, one not → False (structure already differs).
        - Leaf return   : falls out of the general case — both children are
                          (None, None) pairs, both return True, leaf returns True.
        - Internal node : values must match, then BOTH subtree pairs must match.
                          Short-circuit `and` means we bail as soon as one fails.
        - Accumulator   : none — answer flows cleanly through the return value.

        Pairing rule: (p.left, q.left) and (p.right, q.right)
        — "same side must match same side."

        Time : O(min(n, m)) — stops at first mismatch
        Space: O(min(h1, h2)) — recursion stack
        """
        if p is None and q is None:
            return True

        if p is None or q is None:
            return False

        if p.val != q.val:
            return True

        return TreeComparisonProblem.is_same(
            p=p.left, q=q.left
        ) and TreeComparisonProblem.is_same(p=p.right, q=q.right)

    @staticmethod
    def is_symmetric(root: TreeNode | None) -> bool:
        """
        Return True if the tree is a mirror image of itself (symmetric around
        its vertical centre axis).

        Design decision:
            Do NOT try to write this as a single-argument function — it becomes
            awkward immediately. Instead, define a two-argument helper is_mirror(p, q)
            that asks "is subtree p the mirror image of subtree q?"
            Then is_symmetric is simply: is_mirror(root.left, root.right).

            Comparing is_same vs is_mirror side-by-side:
                is_same  pairs (p.left, q.left)  and (p.right, q.right)  — same sides
                is_mirror pairs (p.left, q.right) and (p.right, q.left)  — opposite sides
            That single pairing swap is the ENTIRE difference between the two problems.

        Time : O(n) — every node visited once
        Space: O(h) — recursion stack
        """

        def is_mirror(p: TreeNode | None, q: TreeNode | None) -> bool:
            """
            Design questions:
            - None return   : both None → True (symmetric empty arms).
                              One None, one not → False (arms differ in structure).
            - Leaf return   : both leaves with matching values → True via general case.
            - Internal node : values must match, then cross-pair the children.
            - Accumulator   : none.
            """
            if p is None and q is None:
                return True

            if p is None or q is None:
                return False

            if p.val != q.val:
                return False

            return is_mirror(p=p.left, q=q.right) and is_mirror(p=p.right, q=q.left)

        if root is None:
            return True

        return is_mirror(p=root.left, q=root.right)

    @staticmethod
    def is_subtree(main: TreeNode | None, sub: TreeNode | None) -> bool:
        """
        Return True if sub is a subtree of main — i.e., some node in main
        roots a subtree identical to sub.

        Contract on None:
            sub = None  → True  (empty tree is a subtree of any tree, by convention)
            main = None, sub not None → False (nothing left to search)

        Design:
            Walk main with a single DFS. At every node, ask is_same(node, sub).
            If that returns True anywhere, we're done.

            This is NOT the same recursion as is_same — it's a two-level recursion:
              Outer: DFS over main (looking for a root-match candidate)
              Inner: is_same called at each candidate (verifying the full subtree)

        Complexity:
            Naive  — O(n x m): at each of n nodes, is_same may do O(m) work.
            Better — O(n + m): serialize both trees (bullet 7) and check if
                     serialize(sub) is a substring of serialize(main). Worth
                     knowing this connection exists even if we don't implement it here.

        Time : O(n x m) this implementation
        Space: O(h_main + h_sub) — two stacks, one per recursion level
        """
        if sub is None:
            return True

        if main is None:
            return False

        if TreeComparisonProblem.is_same(p=main, q=sub):
            return True

        return TreeComparisonProblem.is_same(
            p=main.left, q=sub
        ) or TreeComparisonProblem.is_same(p=main.right, q=sub)

    @staticmethod
    def mirror(root: TreeNode | None) -> TreeNode:
        """
        Mutate the tree in-place into its own mirror image by swapping every
        node's left and right children. Return the (same) root.

        Returning root makes the function composable:
            result = mirror(build_tree())
        even though mutation is the primary effect.

        Design questions:
        - None return   : None — nothing to swap, propagate None upward.
        - Leaf return   : swap children (both None↔None, a no-op), return leaf.
                          Falls out of the general case naturally.
        - Internal node : swap left↔right FIRST, then recurse into both children.
                          Order of swap vs recurse doesn't affect correctness, but
                          swap-first is slightly more intuitive: "fix this node,
                          then fix both subtrees."
        - Accumulator   : none — mutation is the side effect, root flows through.

        Time : O(n) — every node visited once
        Space: O(h) — recursion stack
        """
        if root is None:
            return None

        root.left, root.right = root.right, root.left

        TreeComparisonProblem.mirror(root=root.left)
        TreeComparisonProblem.mirror(root=root.right)

        return root


if __name__ == "__main__":
    A = TreeNode(1)
    B = TreeNode(2)
    C = TreeNode(3)
    D = TreeNode(4)
    E = TreeNode(5)
    F = TreeNode(6)
    G = TreeNode(7)
    H = TreeNode(8)

    A.left = B
    A.right = C
    B.left = D
    B.right = E
    D.left = G
    D.right = H
    A.right = C
    C.right = F

    I = TreeNode(1)
    J = TreeNode(2)
    K = TreeNode(3)
    L = TreeNode(4)
    M = TreeNode(5)
    N = TreeNode(6)
    O = TreeNode(7)
    P = TreeNode(8)

    I.left = J
    I.right = K
    J.left = L
    J.right = M
    L.left = O
    L.right = P
    I.right = K
    K.right = N

    # This is what the above tree would look like:
    #        A
    #       / \
    #      B  C
    #     / \  \
    #    D  E  F
    #   / \
    #  G  H

    problem = TreeComparisonProblem()
    print(problem.is_same(p=A, q=I))
    print(problem.is_subtree(main=A, sub=H))

    root = TreeNode(1)
    root.left = TreeNode(2)
    root.right = TreeNode(2)
    root.left.left = TreeNode(3)
    root.left.right = TreeNode(4)
    root.right.left = TreeNode(4)
    root.right.right = TreeNode(3)

    # This is what the above tree would look like:
    #      1
    #     / \\
    #    2   2
    #   / \\ / \\
    #  3  4 4  3
    print(problem.is_symmetric(root=root))
````

### Views and Traversal Variants

This section may have more named problems than others, but there are just two main
mechanism:

- **Mechanism A**: BFS with a bucket key. Do a level-order traversal, and at every node,
  decide which bucket it belongs to based on some key. The key is either:
    - **Its depth/level**: Gives you left/right/top/bottom view problems.
    - **Its column index**: Gives you vertical order, top view, bottom view.
- **Mechanism B**: DFS with careful boundary tracking. Walk specific structural
  boundaries of the tree in a defined order. You're not visiting every node here, you're
  tracing the outline of the tree.

#### Left and Right Views

Set of nodes visible when you look at the tree from left or right. Basically, the first
left or right nodes in each level.

#### Top and Bottom Views

Now you switch the bucket key from depth to column index. You assign a column number to
every node. Root column's is 0, going left decrements, and going right increments.

```txt
         1  (col 0)
        / \
       2   3    (col -1, col 1)
      / \   \
     4   5   6  (col -2, col 0, col 2)
```

Top View: -2→4, -1→2, 0→1, 1→3, 2→6 → [4, 2, 1, 3, 6]
Bottom View: -2→4, -1→2, 0→5 (5 overwrites 1), 1→3, 2→6 → [4, 2, 5, 3, 6]

In top view, `1` wins for column 0, but in the bottom view, `5` wins.

#### Vertical Order Traversal

This is the full version of column-bucket idea, instead of keeping one node at each
column, collect all nodes, grouped by columns, and within the column groups, the nodes
are ordered by the depth and the left-to-right order.

#### Zigzag/Spiral Traversal

Standard level-order BFS, but change the direction when you go deeper.

Level 0 -> left to right. Level 1 -> right to left. Level 2 -> left to right ...

#### Boundary Traversal

This traversal uses the second mechanism. In this model, boundary of a tree is its
"outline": left side goes down, then all the leaves, then right side going back up.

There are three steps on the boundary traversal:

- First, you should go left down, from root to the deepest left node, whenever is
  possible you should go left, and if there is no left, you should go right.
- Then, you should visit all the leafs, from left to right.
- At last, there are right boundary nodes, from the deepest right node, to the root.  

````python
from collections import defaultdict, deque
from typing import NamedTuple, Self


class TreeNode:
    """Represent a node in a binary tree."""

    def __init__(
        self,
        val: int = 0,
        left: Self | None = None,
        right: Self | None = None,
    ) -> None:
        """Intialize a tree node."""
        self.val = val
        self.left = left
        self.right = right

    def __str__(self) -> str:
        """Return the string representation of the class."""
        return f"TreeNode <{self.val}>"

    def __repr__(self) -> str:
        """Return the official string representation of the class."""
        return f"TreeNode({self.val})"


class ViewPair(NamedTuple):
    """Tuple representing a pair in the queue at top and bottom view operations."""

    column: int
    node: TreeNode


class TreeTraversal:
    """
    All view and traversal-variant operations on a plain binary tree.

    Naming convention for BFS queues:
        Items are always (node, key) tuples.
        Key is 'col' for column problems, 'depth' for depth problems,
        or absent (just node) when the level-size loop gives enough info.
    """

    @staticmethod
    def show_left_view(root: TreeNode | None) -> list[int]:
        """
        Return the first node at each depth level — the nodes visible
        when looking at the tree from the left side.

        Mechanic A: BFS, bucket key = depth (implicit via level-size loop).
        What to keep: FIRST node encountered per level.

        The level-size trick:
            Before processing a level, snapshot len(queue). That count tells
            you exactly how many nodes belong to the current level. Process
            that many, then whatever remains in the queue is the next level.
            This avoids carrying an explicit depth in the queue.

        Design:
        - Return value: list of node values (one per level)
        - Accumulator : result list, appended as side effect per level

        Time : O(n)
        Space: O(w) — w = max width of the tree (queue holds one level at a time)
        """
        if root is None:
            return []

        result: list[int] = []
        queue: deque[TreeNode] = deque([root])

        while queue:
            level_size = len(queue)
            for i in range(level_size):
                node = queue.popleft()
                if i == 0:
                    result.append(node.val)

                if node.left is not None:
                    queue.append(node.left)
                if node.right is not None:
                    queue.append(node.right)

        return result

    @staticmethod
    def show_right_view(root: TreeNode | None) -> list[int]:
        """
        Return the last node at each depth level — the nodes visible
        when looking at the tree from the right side.

        Mechanic A: BFS, bucket key = depth (implicit via level-size loop).
        What to keep: LAST node encountered per level.

        Identical structure to left_view — the only change is `i == 0`
        becomes `i == level_size - 1`.

        Time : O(n)
        Space: O(w)
        """
        if root is None:
            return []

        result: list[int] = []
        queue: deque[TreeNode] = deque([root])

        while queue:
            level_size = len(queue)
            for i in range(level_size):
                node = queue.popleft()
                if i == level_size - 1:
                    result.append(node.val)

                if node.left is not None:
                    queue.append(node.left)
                if node.right is not None:
                    queue.append(node.right)

        return result

    @staticmethod
    def show_top_view(root: TreeNode | None) -> list[int]:
        """
        Return the first (topmost) node in each column, ordered left to right.
        A node is "visible from the top" if no ancestor sits in the same column.

        Mechanic A: BFS, bucket key = column index.
        What to keep: FIRST node per column (BFS visits top-to-bottom,
        so "first seen" is always the topmost).

        Column assignment:
            root → col 0
            any node at col c: left child → col c-1, right child → col c+1

        The queue carries (node, col) pairs.

        Design:
        - col_map[col] = value of topmost node in that column
        - After BFS, sort col_map keys and read values in order

        Time : O(n log n) — due to sorting column keys at the end
        Space: O(n) — queue and col_map together
        """
        if root is None:
            return []

        column_map: dict[int, int] = {}
        queue: deque[ViewPair] = deque([ViewPair(column=0, node=root)])

        while queue:
            column, node = queue.popleft()

            if column not in column_map:
                column_map[column] = node.val

            if node.left is not None:
                queue.append(ViewPair(column=column - 1, node=node.left))
            if node.right is not None:
                queue.append(ViewPair(column=column + 1, node=node.right))

        return [column_map[column] for column in sorted(column_map)]

    @staticmethod
    def show_bottom_view(root: TreeNode | None) -> list[int]:
        """
        Return the last (bottommost) node in each column, ordered left to right.

        Mechanic A: BFS, bucket key = column index.
        What to keep: LAST node per column (always overwrite — BFS visits
        top-to-bottom, so whatever was written last is the deepest node.
        When two nodes share the same column and depth, the one encountered
        later in BFS — i.e., the rightmost — wins, which is the expected
        LeetCode/GFG convention for ties).

        Structurally identical to top_view; the only change:
            top_view  → write if col NOT yet in col_map  (first = topmost)
            bottom_view → ALWAYS overwrite               (last  = bottommost)

        Time : O(n log n)
        Space: O(n)
        """

        if root is None:
            return []

        column_map: dict[int, int] = {}
        queue: deque[ViewPair] = deque([ViewPair(column=0, node=root)])

        while queue:
            column, node = queue.popleft()

            column_map[column] = node.val

            if node.left is not None:
                queue.append(ViewPair(column=column - 1, node=node.left))
            if node.right is not None:
                queue.append(ViewPair(column=column + 1, node=node.right))

        return [column_map[column] for column in sorted(column_map)]

    @staticmethod
    def show_vertical_view(root: TreeNode | None) -> list[list[int]]:
        """
        Group ALL nodes by column (left to right). Within each column, nodes
        appear top-to-bottom; ties at the same (col, depth) appear
        left-to-right (i.e., in BFS encounter order).

        Mechanic A: BFS, bucket key = column index.
        What to keep: ALL nodes per column, in BFS order (which gives
        top-to-bottom and left-to-right for free — no explicit depth needed).

        This is the fullest version of the column-bucket idea:
            top_view    → first  per column
            bottom_view → last   per column
            vertical    → ALL    per column, BFS order

        Design:
        - col_map: defaultdict mapping col → list of vals (appended in BFS order)
        - After BFS, sort keys and return the lists

        Time : O(n log n)
        Space: O(n)
        """
        if root is None:
            return []

        column_map: defaultdict[int, list[int]] = defaultdict(list)
        queue: deque[ViewPair] = deque([ViewPair(column=0, node=root)])

        while queue:
            column, node = queue.popleft()

            column_map[column].append(node.val)

            if node.left is not None:
                queue.append(ViewPair(column=column - 1, node=node.left))
            if node.right is not None:
                queue.append(ViewPair(column=column + 1, node=node.right))

        return [column_map[column] for column in sorted(column_map)]

    @staticmethod
    def show_zigzag_view(root: TreeNode | None) -> list[list[int]]:
        """
        Level-order traversal, but alternate collection direction each level:
            Level 0 (root): left → right
            Level 1       : right → left
            Level 2       : left → right  ...

        Mechanic: BFS with a direction flag that flips after each level.

        Two implementation approaches exist:
          A) Build level list normally, then reverse odd levels. (used here)
             Simpler to reason about, one extra O(w) reverse per odd level.
          B) Use a deque for level collection; alternate appendleft/append.
             Avoids reversals but the logic is trickier to read.

        We use approach A — the reversal cost is negligible vs. clarity gained.

        Design:
        - left_to_right flag starts True, flips after each level
        - level list built normally during BFS, reversed if needed before appending

        Time : O(n)
        Space: O(w)
        """
        if root is None:
            return []

        result: list[list[int]] = []
        queue: deque[TreeNode] = deque([root])
        left_to_right = True

        while queue:
            level = []
            level_size = len(queue)

            for _ in range(level_size):
                node = queue.popleft()
                level.append(node.val)

                if node.left is not None:
                    queue.append(node.left)
                if node.right is not None:
                    queue.append(node.right)

            if not left_to_right:
                level.reverse()
            result.append(level)
            left_to_right = not left_to_right

        return result

    @staticmethod
    def show_boundary_traversal(root: TreeNode | None) -> list[int]:
        """
        Return the boundary (outline) of the tree as a list of node values
        visited anti-clockwise: root → left boundary → leaves → right boundary.

        This is the odd one out — NOT a simple BFS bucket problem.
        The boundary is defined as three non-overlapping pieces:

          Piece 1 — Root (just root.val)

          Piece 2 — Left boundary (top-down, EXCLUDING leaves):
              Start at root.left. Go left when possible; if no left child,
              go right. Stop before reaching a leaf (to avoid double-counting
              it in piece 3).

          Piece 3 — Leaves (left-to-right, entire tree via DFS):
              Every node with no children, in left-to-right order.
              Includes leaves on the left AND right boundary — they're
              collected here, not in pieces 2 or 4.

          Piece 4 — Right boundary (bottom-up, EXCLUDING leaves and root):
              Start at root.right. Go right when possible; if no right child,
              go left. Stop before leaves. Collect in reverse (append AFTER
              recursing) so the output is bottom-up not top-down.

        Why three helpers?
            Each piece has a subtly different walk rule. Collapsing them into
            one function leads to messy flag-passing. Three focused helpers are
            clearer and easier to test in isolation.

        Edge cases handled:
          - None root → []
          - Single node (leaf) → [root.val] — no left/right boundary to walk
          - Left-only or right-only trees — the absent side simply contributes []

        Time : O(n) — every node visited at most once across all three helpers
        Space: O(h) — recursion stack depth
        """
        if root is None:
            return []

        result: list[int] = [root.val]

        if root.left is None and root.right is None:
            return result

        def add_left_boundary(node: TreeNode | None) -> None:
            """Walk down the left boundary top-to-bottom, skipping leaves."""
            if node is None or (node.left is None and node.right is None):
                return None

            result.append(node.val)

            if node.left:
                add_left_boundary(node=node.left)
            else:
                add_left_boundary(node=node.right)

        def add_leaves(node: TreeNode | None) -> None:
            """Collect all leaves left-to-right via DFS."""
            if node is None:
                return None
            if node.left is None and node.right is None:
                result.append(node.val)
                return
            add_leaves(node.left)
            add_leaves(node.right)

        def add_right_boundary(node: TreeNode | None) -> None:
            """Walk down the right boundary then collect bottom-up, skipping leaves."""
            if node is None or (node.left is None and node.right is None):
                return None

            if node.right:
                add_right_boundary(node=node.right)
            else:
                add_right_boundary(node=node.left)

            result.append(node.val)

        add_left_boundary(node=root.left)
        add_leaves(root)
        add_right_boundary(node=root.right)
        return result


if __name__ == "__main__":
    A = TreeNode(1)
    B = TreeNode(2)
    C = TreeNode(3)
    D = TreeNode(4)
    E = TreeNode(5)
    F = TreeNode(6)
    G = TreeNode(7)
    H = TreeNode(8)

    A.left = B
    A.right = C
    B.left = D
    B.right = E
    D.left = G
    D.right = H
    A.right = C
    C.right = F

    # This is what the above tree would look like:
    #        1
    #       / \
    #      2  3
    #     / \  \
    #    4  5  6
    #   / \
    #  7  8

    traversal = TreeTraversal()
    print(traversal.show_left_view(root=A))
    print(traversal.show_right_view(root=A))
    print(traversal.show_top_view(root=A))
    print(traversal.show_bottom_view(root=A))
    print(traversal.show_vertical_view(root=A))
    print(traversal.show_zigzag_view(root=A))
    print(traversal.show_boundary_traversal(root=A))
````

### Reconstruction from Traversal Pairs

I already know how to produce traversals from a tree, here, we flips the problem: **How
should I uniquely reconstruct a tree from one or two given traversal sequences.

#### None of each DFS Versions or BFS Traversals Can Lead Us to Reconstruction Alone

preorder, postorder, inorder, and level-order traversals can not lead us to tree
reconstruction alone, but it's worth mentioning that `preorder` and `inorder` are
complementary. Preorder identifies the root; inorder splits the remaining elements into
left and right subtrees. Together they can encode the full structure.

#### preorder + inorder Reconstruction

```txt
Tree:
        3
       / \
      9  20
        /  \
       15   7

Preorder: [3, 9, 20, 15, 7]
Inorder:  [9, 3, 15, 20, 7]
```

- **Step 1:** Preorder's first element is 3, so 3 is the root.
- **Step 2:** Find 3 in inorder. Everything to left is the left subtree, and everything
  to its right is the right subtree.
- **Step 3:** Left subtree has one element, so the next one element in preorder is the
  left subtree's preorder. The next three elements are the right subtree's preorder.
- **Step 4:** Recurse on:
  - **Left:** preorder `[9]`, inorder `[9]` -> root is `9`, no children.
  - **Right:** preorder `[20, 15, 7]`, inorder `[15, 20, 7]` -> roots is `20`, left
    is `15`, right is `7`.

> 🖊 The numbers of elements in the left subtree in inorder tells you exactly how many
> elements to take from preorder for the left subtree.

#### The O(n^2) Trap and the O(n) Fix

The naive's implementation searches for the root's index in the inorder array with a
linear scan at every recursive call. For a balanced tree iis O(nlogn), for a skewed tree
it's O(n^2).

For fixing the issue, you should precompute a `value -> index` dictionary for the
inorder array before recursion starts. Then every root lookup is O(1), and the whole
reconstruction is O(n).

#### postorder + inorder Reconstruction

Same problem, just the root is at the end of the postorder list.

````python
from typing import Self


class TreeNode:
    """Represent a node in a binary tree."""

    def __init__(
        self,
        val: int = 0,
        left: Self | None = None,
        right: Self | None = None,
    ) -> None:
        """Initialize a tree node."""
        self.val = val
        self.left = left
        self.right = right

    def __str__(self) -> str:
        """Return the string representation of the class."""
        return f"TreeNode <{self.val}>"

    def __repr__(self) -> str:
        """Return the official string representation of the class."""
        return f"TreeNode({self.val})"


def preorder(root: TreeNode | None) -> list[int]:
    """Root → Left → Right."""
    if root is None:
        return []
    return [root.val] + preorder(root.left) + preorder(root.right)


def inorder(root: TreeNode | None) -> list[int]:
    """Left → Root → Right."""
    if root is None:
        return []
    return inorder(root.left) + [root.val] + inorder(root.right)


def postorder(root: TreeNode | None) -> list[int]:
    """Left → Right → Root."""
    if root is None:
        return []
    return postorder(root.left) + postorder(root.right) + [root.val]


class TreeReconstruction:
    """
    Reconstruct a binary tree from pairs of traversal sequences.

    Assumption (required by both methods):
        All node values are UNIQUE. If duplicates existed, the
        value → index mapping in inorder would be ambiguous and
        the reconstruction would not be well-defined.
    """

    @staticmethod
    def build_from_preorder_inorder(
        preorder_seq: list[int],
        inorder_seq: list[int],
    ) -> TreeNode | None:
        """
        Reconstruct a binary tree from its preorder and inorder traversals.

        Core mechanic:
            preorder[preorder_start_index]  → root of the current subtree (always)
            inorder_index[root]  → position that splits inorder into left/right
            left_size            → how many nodes are in the left subtree
                                   (same count in both preorder and inorder slices)

        Index arithmetic (all bounds are inclusive):
            preorder_start_index                         → root of this subtree
            [preorder_start_index+1 .. preorder_start_index+left_size]  → left subtree in preorder
            [preorder_start_index+left_size+1 .. preorder_end_index]    → right subtree in preorder

            [inorder_start_index .. in_root-1]           → left subtree in inorder
            in_root                           → root (skip it)
            [in_root+1 .. inorder_end_index]             → right subtree in inorder

        Why index bounds instead of slices?
            list[a:b] creates a NEW list — O(n) time and space per call.
            With n recursive calls, that's O(n²) total space.
            Passing integer indices into the original arrays keeps space O(n)
            (just the recursion stack) and time O(n) (each node visited once).

        Prerequisite:
            inorder_index dict is built ONCE before recursion — O(n) upfront,
            O(1) per lookup. Without it, each lookup is O(n) → O(n²) total.

        Time : O(n)
        Space: O(n) — inorder_index dict + O(h) recursion stack
        """
        if not preorder_seq or not inorder_seq:
            return None

        inorder_index: dict[int, int] = {
            value: index
            for index, value in enumerate(inorder_seq)
        }

        def build(
            preorder_start_index: int,
            preorder_end_index: int,
            inorder_start_index: int,
            inorder_end_index: int,
        ) -> TreeNode | None:
            if preorder_start_index > preorder_end_index:
                return None

            root_value = preorder_seq[preorder_start_index]
            root = TreeNode(val=root_value)

            root_inorder_index = inorder_index[root_value]
            left_size = root_inorder_index - inorder_start_index

            root.left = build(
                preorder_start_index=preorder_start_index + 1,
                preorder_end_index=preorder_start_index + left_size,
                inorder_start_index=inorder_start_index,
                inorder_end_index=root_inorder_index - 1,
            )

            root.right = build(
                preorder_start_index=preorder_start_index + 1 + left_size,
                preorder_end_index=preorder_end_index,
                inorder_start_index=root_inorder_index + 1,
                inorder_end_index=inorder_end_index,
            )

            return root

        return build(0, len(preorder_seq) - 1, 0, len(inorder_seq) - 1)

    @staticmethod
    def build_from_postorder_inorder(
        postorder_seq: list[int],
        inorder_seq: list[int],
    ) -> TreeNode | None:
        """Reconstruct a binary tree from its postorder and inorder traversals."""
        if not postorder_seq or not inorder_seq:
            return None

        inorder_index: dict[int, int] = {
            value: index
            for index, value in enumerate(inorder_seq)
        }

        def build(
            postorder_start_index: int,
            postorder_end_index: int,
            inorder_start_index: int,
            inorder_end_index: int,
        ) -> TreeNode | None:
            if postorder_start_index > postorder_end_index:
                return None

            root_value = postorder_seq[postorder_end_index]
            root = TreeNode(val=root_value)

            inorder_root_index = inorder_index[root_value]
            right_size = inorder_end_index - inorder_root_index

            root.left = build(
                postorder_start_index=postorder_start_index,
                postorder_end_index=postorder_end_index - right_size - 1,
                inorder_start_index=inorder_start_index,
                inorder_end_index=inorder_root_index - 1,
            )

            root.right = build(
                postorder_start_index=postorder_end_index - right_size,
                postorder_end_index=postorder_end_index - 1,
                inorder_start_index=inorder_root_index + 1,
                inorder_end_index=inorder_end_index,
            )

            return root

        return build(
            postorder_start_index=0,
            postorder_end_index=len(postorder_seq) - 1,
            inorder_start_index=0,
            inorder_end_index=len(inorder_seq) - 1,
        )


if __name__ == "__main__":
    A = TreeNode(1)
    B = TreeNode(2)
    C = TreeNode(3)
    D = TreeNode(4)
    E = TreeNode(5)
    F = TreeNode(6)
    G = TreeNode(7)
    H = TreeNode(8)

    A.left = B
    A.right = C
    B.left = D
    B.right = E
    D.left = G
    D.right = H
    A.right = C
    C.right = F

    # This is what the above tree would look like:
    #        1
    #       / \
    #      2  3
    #     / \  \
    #    4  5  6
    #   / \
    #  7  8

    reconstruction = TreeReconstruction()
    root = reconstruction.build_from_preorder_inorder(
        preorder_seq=preorder(root=A),
        inorder_seq=inorder(root=A),
    )
    print("================ preorder + inorder ================")
    print(root)
    print(root.left)
    print(root.right)
    print(root.left.left)
    print(root.left.right)
    print(root.left.left.left)
    print(root.left.left.right)
    print(root.right.right)

    root = reconstruction.build_from_postorder_inorder(
        postorder_seq=postorder(root=A),
        inorder_seq=inorder(root=A),
    )
    print("================ postorder + inorder ================")
    print(root)
    print(root.left)
    print(root.right)
    print(root.left.left)
    print(root.left.right)
    print(root.left.left.left)
    print(root.left.left.right)
    print(root.right.right)
````

### Serialization/Deserialization

Serialization is converting a tree from its in-memory form into a flat string that can
be stored to disk, sent over a network, or embedded in another data structure.

Deserialization is exactly the reverse. Key question is: **What does serialization
format need to contain so that deserialization can reconstruct the tree exactly?**

From the previous chapter you know that different tree structures can result into the
same preorder, postorder, and level-order sequences. The solution is *marking `None`
children explicitly*.

#### Two Approaches

**Approach 1**: DFS (preorder) with null markers, the most natural recursive design.
Serialize mirrors the DFS recursion exactly: emit the current node's value, recurse left
and then recurse right. Deserialization replays the same DFS order, consuming tokens one
at a time from an iterator.

**Approach 2**: BFS with null markers, more visual output. The output matches the level
by level shape you'd draw on paper.

#### Subtree O(n + m) Check

Serialize both threes, then check if the serialized sub is a sub-string of main. This
would be done in O(n + m) instead of O(n x m).

````python
from collections import deque
from typing import Self


class TreeNode:
    """Represent a node in a binary tree."""

    def __init__(
        self,
        val: int = 0,
        left: Self | None = None,
        right: Self | None = None,
    ) -> None:
        """Initialize a tree node."""
        self.val = val
        self.left = left
        self.right = right

    def __str__(self) -> str:
        """Return the string representation of the class."""
        return f"TreeNode <{self.val}>"

    def __repr__(self) -> str:
        """Return the official string representation of the class."""
        return f"TreeNode({self.val})"


class TreeDFSSerialization:
    r"""
    Serialize / deserialize using preorder DFS with explicit null markers.

    Serialized format example:
        Tree:      3
                  / \\
                 9  20
                   /  \\
                  15   7

        Output: "3,9,#,#,20,15,#,#,7,#,#"

    Why preorder?
        Preorder visits root FIRST — so the first token is always the root.
        This mirrors the recursive structure: serialize(node) = node's value,
        then serialize(left subtree), then serialize(right subtree).
        Deserialization replays this exact order.

    The iterator trick:
        Instead of tracking an index into the token list across recursive
        calls, we wrap the token list in an iterator. Each call to next()
        consumes exactly one token and automatically advances the position.
        The iterator state is shared across all recursive calls — no index
        management needed. This is the cleanest way to write recursive
        deserialization.
    """

    NONE_MARKER = "#"
    SEPARATOR = ","

    @classmethod
    def serialize(cls, root: TreeNode | None) -> str:
        """
        Preorder DFS: emit node value, recurse left, recurse right.
        Emit `NONE_MARKER` when a None is encountered.

        Design:
        - tokens list is built as a side effect of the DFS.
        - No return value from the inner helper — tokens accumulates everything.
        - Base case: node is None → append '#', return immediately.

        Time : O(n) — every node (and null slot) visited exactly once
        Space: O(n) — tokens list + O(h) recursion stack
        """
        tokens: list[str] = []

        def preorder(node: TreeNode | None) -> None:
            if node is None:
                tokens.append(cls.NONE_MARKER)
                return
            tokens.append(str(node.val))
            preorder(node=node.left)
            preorder(node=node.right)

        preorder(node=root)
        return cls.SEPARATOR.join(tokens)

    @classmethod
    def deserialize(cls, data: str) -> TreeNode | None:
        """
        Replay the same preorder DFS, consuming one token per node/null.

        The iterator pattern:
            token_iter = iter(data.split(','))
            Each recursive build() call consumes its own token via next(),
            then its left child's build() consumes the next, and so on.
            The iterator advances automatically — no index arithmetic.

        Design:
        - Base case: token == '#' → return None (null slot in the original tree).
        - Each non-null token creates a TreeNode whose children are built by
          two immediate recursive calls, consuming the next tokens in order.
        - The recursion bottoms out at '#' markers, exactly mirroring how
          serialize() emitted them.

        Time : O(n)
        Space: O(n) token list + O(h) recursion stack
        """
        token_iter = iter(data.split(cls.SEPARATOR))

        def build() -> TreeNode | None:
            value = next(token_iter)
            if value == cls.NONE_MARKER:
                return None

            root = TreeNode(val=value)
            root.left = build()
            root.right = build()
            return root

        return build()


class TreeBFSSerialization:
    """
    Serialize / deserialize using level-order BFS with explicit null markers.

    Serialized format example (same tree as above):
        Output: "3,9,20,#,#,15,7,#,#,#,#"

    This is the format LeetCode uses in its tree visualizer, making it
    intuitive to read — values appear left-to-right, level by level.

    Serialize:
        Standard BFS with a queue. Enqueue None children too — emit '#'
        for them. This ensures the null structure is encoded.
        We stop enqueueing children of null nodes (there's nothing to enqueue).

    Deserialize:
        The key insight: every non-null node in the serialized sequence
        has exactly two children (which may be null) occupying the next
        two positions in the token list. Use a queue of "nodes waiting
        for their children" — pop a node, assign its left child from the
        next token, assign its right child from the token after that.
        Only enqueue a child if it's non-null (null nodes have no children
        to assign in future iterations).
    """

    NONE_MARKER = "#"
    SEPARATOR = ","

    @classmethod
    def serialize(cls, root: TreeNode | None) -> str:
        """
        BFS traversal; enqueue None children and emit '#' for them.

        Design:
        - Queue holds TreeNode | None.
        - For every node dequeued: emit its value (or '#'), then enqueue
          both children if the node is non-null.
        - Null nodes are emitted but not expanded (they have no children).

        Time : O(n)
        Space: O(w) — queue holds at most one full level at a time
        """
        if root is None:
            return cls.NONE_MARKER

        tokens: list[str] = []
        queue: deque[TreeNode] = deque([root])

        while queue:
            node = queue.popleft()

            if node is None:
                tokens.append(cls.NONE_MARKER)
            else:
                tokens.append(str(node.val))
                queue.append(node.left)
                queue.append(node.right)

        return cls.SEPARATOR.join(tokens)

    @classmethod
    def deserialize(cls, data: str) -> TreeNode | None:
        """
        Replay BFS: each non-null node consumes the next two tokens as children.

        Design:
        - Parse the root from tokens[0]; if '#', return None immediately.
        - Queue of "nodes awaiting children assignment."
        - Walk token index i from 1 onward, always in steps of 2
          (left child, then right child, for the current queue-front node).
        - Only enqueue a child if it was non-null — null children don't
          need children of their own assigned in future iterations.

        Why not use an iterator here (like DFS)?
            BFS deserialization is inherently index-driven — we step through
            tokens[i] and tokens[i+1] per node. An iterator works but an
            explicit index is equally clean and slightly easier to follow
            since there's no recursion to "naturally advance" it.

        Time : O(n)
        Space: O(w) queue
        """
        tokens = data.split(cls.SEPARATOR)
        if (not tokens) or tokens[0] == cls.SEPARATOR:
            return None

        root = TreeNode(val=int(tokens[0]))
        queue: deque[TreeNode] = deque([root])
        i = 1

        while queue:
            node = queue.popleft()

            if tokens[i] != cls.NONE_MARKER:
                node.left = TreeNode(val=int(tokens[i]))
                queue.append(node.left)
            i += 1

            if tokens[i] != cls.NONE_MARKER:
                node.right = TreeNode(val=int(tokens[i]))
                queue.append(node.right)
            i += 1

        return root


if __name__ == "__main__":
    A = TreeNode(1)
    B = TreeNode(2)
    C = TreeNode(3)
    D = TreeNode(4)
    E = TreeNode(5)
    F = TreeNode(6)
    G = TreeNode(7)
    H = TreeNode(8)

    A.left = B
    A.right = C
    B.left = D
    B.right = E
    D.left = G
    D.right = H
    A.right = C
    C.right = F

    # This is what the above tree would look like:
    #        1
    #       / \
    #      2  3
    #     / \  \
    #    4  5  6
    #   / \
    #  7  8

    serializer = TreeDFSSerialization()
    serialized_tree = serializer.serialize(root=A)
    print("================ DFS Serialization ================")
    print(serialized_tree)

    root = serializer.deserialize(data=serialized_tree)
    print("================ DFS Deserialization ================")
    print(root)
    print(root.left)
    print(root.right)
    print(root.left.left)
    print(root.left.right)
    print(root.left.left.left)
    print(root.left.left.right)
    print(root.right.right)

    serializer = TreeBFSSerialization()
    serialized_tree = serializer.serialize(root=A)
    print("================ BFS Serialization ================")
    print(serialized_tree)

    root = serializer.deserialize(data=serialized_tree)
    print("================ BFS Deserialization ================")
    print(root)
    print(root.left)
    print(root.right)
    print(root.left.left)
    print(root.left.right)
    print(root.left.left.left)
    print(root.left.left.right)
    print(root.right.right)
````

### Morris Traversal

Normal inorder traversal uses O(h) extra space: the call stack in the recursive version,
and an explicit stack in the iterative version. For a balanced tree that's O(logn) but
fr a skewed tree it is O(n). Morris traversal does inorder in O(1) extra space by
temporarily modifying the tree itself, then restoring it.

I find this traversal hard to learn and it was necessary for me to watch
[this video](https://www.youtube.com/watch?v=wGXB9OWhPTg&t=547s) to learn it, so I cut
the explanations. You can watch the video in order to learn the traversal.

````python
from typing import Self


class TreeNode:
    """Represent a node in a binary tree."""

    def __init__(
        self,
        val: int = 0,
        left: Self | None = None,
        right: Self | None = None,
    ) -> None:
        """Initialize a tree node."""
        self.val = val
        self.left = left
        self.right = right

    def __str__(self) -> str:
        """Return the string representation of the class."""
        return f"TreeNode <{self.val}>"

    def __repr__(self) -> str:
        """Return the official string representation of the class."""
        return f"TreeNode({self.val})"


class MorrisTraversal:
    """Inlcude morris traversal inorder and preorder algorithms."""

    @staticmethod
    def traverse_inorder(root: TreeNode | None) -> list[int]:
        """
        Inorder traversal (Left → Root → Right) in O(1) space.

        Algorithm:
            current = root
            while current is not None:
                if current has no left child:
                    ① VISIT current          (leftmost reachable node, or we've
                                              just returned via a thread)
                    ② current = current.right
                else:
                    find inorder predecessor of current
                    (= rightmost node in current.left subtree)

                    if predecessor.right is None:   # no thread yet
                        ③ install thread: predecessor.right = current
                        ④ current = current.left    # descend into left subtree
                    else:                           # thread points back to us
                        ⑤ VISIT current             # left subtree fully done
                        ⑥ remove thread: predecessor.right = None
                        ⑦ current = current.right   # move to right subtree

        Why the inner while loop terminates:
            We walk right from current.left until:
            - predecessor.right is None  (normal unthreaded case), OR
            - predecessor.right is current (thread we installed earlier).
            Both conditions are guaranteed to be reached — None because the
            rightmost node has no right child, and the thread because we set
            predecessor.right = current in the first visit.

        Tree restoration guarantee:
            Every thread installed in step ③ is removed in step ⑥ before
            current moves past that node. When the while loop exits, the
            tree is structurally identical to its state before the call.
        """
        result: list[int] = []
        current = root

        while current is not None:
            # No left subtree - this node should be visited in natural order. Either
            # it's a leaf or we've just followed a thread back here and already
            # exhausted the left subtree.
            if current.left is None:
                result.append(current.val)
                current = current.right
            else:
                # Find the predecessor - the rightmost node reachable from current.left.
                predecessor = current.left
                while (
                    predecessor.right is not None and predecessor.right is not current
                ):
                    predecessor = predecessor.right

                if predecessor.right is None:
                    # First encounter and there are no threads yet. Thread the
                    # predecessor to the current, so we can return back here after
                    # exhuasting the left subtree.
                    predecessor.right = current
                    current = current.left
                else:
                    # Second encounter, meaning the predecessor right is current,
                    # meaning we have fully return from an exhausted left subtree.
                    predecessor.right = None
                    result.append(current.val)
                    current = current.right
        return result

    @staticmethod
    def traverse_preorder(root: TreeNode) -> list[int]:
        """
        Preorder traversal (Root → Left → Right) in O(1) space.

        Identical to the inorder algorithm with ONE change:
            Visit current on the FIRST encounter (when installing the thread),
            not the second. When the thread brings us back, just remove it and
            move right — the node was already visited.

        Nodes with no left child are identical in both variants:
            visit, then move right.

        Diff from inorder (marked with ←):
            if predecessor.right is None:
                result.append(current.val)    # ← visit HERE (first encounter)
                predecessor.right = current
                current = current.left
            else:
                predecessor.right = None
                # (no visit here)             # ← no visit on second encounter
                current = current.right
        """
        result: list[int] = []
        current = root

        while current is not None:
            if current.left is None:
                result.append(current.val)
                current = current.right
            else:
                predecessor = current.left
                while (
                    predecessor.right is not None and predecessor.right is not current
                ):
                    predecessor = predecessor.right

                if predecessor.right is None:
                    result.append(current.val)
                    predecessor.right = current
                    current = current.left
                else:
                    predecessor.right = None
                    current = current.right
        return result


if __name__ == "__main__":
    A = TreeNode(1)
    B = TreeNode(2)
    C = TreeNode(3)
    D = TreeNode(4)
    E = TreeNode(5)
    F = TreeNode(6)
    G = TreeNode(7)
    H = TreeNode(8)

    A.left = B
    A.right = C
    B.left = D
    B.right = E
    D.left = G
    D.right = H
    A.right = C
    C.right = F

    # This is what the above tree would look like:
    #        1
    #       / \
    #      2  3
    #     / \  \
    #    4  5  6
    #   / \
    #  7  8

    traversal = MorrisTraversal()
    print(traversal.traverse_inorder(root=A))
    print(traversal.traverse_preorder(root=A))
````

### LCA in Plain Binary Tree

**What is LCA?** The lowest common ancestor of two nodes `p` and `q` is the deepest node
in the tree that has both `p` and `q` as descendants, where a node is considered a
descendant of itself. "Deepest" is what makes it lowest (furthest from root).

This problem is easier to solve in binary search tree because of its ordering, and it's
harder in plain binary trees that naturally has no order.

````python
from collections import deque
from typing import Self


class TreeNode:
    """Represent a node in a binary tree."""

    def __init__(
        self,
        val: int = 0,
        left: Self | None = None,
        right: Self | None = None,
    ) -> None:
        """Intialize a tree node."""
        self.val = val
        self.left = left
        self.right = right

    def __str__(self) -> str:
        """Return the string representation of the class."""
        return f"TreeNode <{self.val}>"

    def __repr__(self) -> str:
        """Return the official string representation of the class."""
        return f"TreeNode({self.val})"


class TreeLCA:
    @staticmethod
    def find(root: TreeNode | None, p: TreeNode, q: TreeNode) -> TreeNode | None:
        """
        Recursive LCA using a triple-duty return value.

        Return value semantics (three simultaneous meanings):
            None        → neither p nor q exists in this subtree
            p or q      → one target was found; the other is either
                          its descendant (making it the LCA) or
                          lives in a sibling subtree (keep searching up)
            another node→ this IS the resolved LCA; both targets were
                          found split across the two subtrees

        Base cases:
            root is None    → empty subtree, return None
            root is p or q  → found a target, return it immediately

        Recursive step:
            left  = find(root.left,  p, q)
            right = find(root.right, p, q)

            Both non-None  → p and q are in different subtrees
                             → root is the split point = LCA → return root
            One non-None   → propagate it upward; the caller will resolve
            Both None      → neither target in this subtree → return None

        Design questions (from bullet 2):
        ┌──────────────┬─────────────────────────────────────────────────┐
        │ Base case    │ None → None; root is p/q → return root          │
        │ Leaf         │ leaf is p or q → leaf; else None                │
        │ Internal     │ check left+right, return root / propagate / None │
        │ Accumulator  │ none — LCA surfaces entirely through return vals │
        └──────────────┴─────────────────────────────────────────────────┘

        Time : O(n)  — every node visited at most once
        Space: O(h)  — recursion stack; O(log n) balanced, O(n) skewed
        """
        if root is None:
            return None

        if root is p or root is q:
            return root

        left_result = TreeLCA.find(root=root.left, p=p, q=q)
        right_result = TreeLCA.find(root=root.right, p=p, q=q)

        if left_result is not None and right_result is not None:
            return root

        return left_result if left_result is not None else right_result

    @staticmethod
    def find_iterative(
        root: TreeNode | None,
        p: TreeNode,
        q: TreeNode,
    ) -> TreeNode | None:
        """
        Iterative LCA via parent-pointer map + ancestor set.

        Intuition:
            If you could walk upward from both p and q simultaneously,
            the first node they share on those paths is the LCA.
            Parent pointers make upward walks possible.

        Steps:
            1. BFS the entire tree to build:
                   parent: dict[TreeNode, TreeNode | None]
               where parent[node] is node's parent (None for root).

            2. Walk up from p, collecting every ancestor of p
               (including p itself) into p_ancestors set.

            3. Walk up from q. The first node encountered that appears
               in p_ancestors is the LCA.

        Why step 3 terminates:
            Both p and q are in the tree (by assumption). Their paths
            to root converge at the LCA at the latest at the root itself.
            So we will always hit a node in p_ancestors before or at root.

        When to prefer this over recursive:
            - The tree is very deep and you want to avoid stack overflow.
            - You need to answer multiple LCA queries on the same tree
              (build the parent map once; reuse for each query).
            - You find the iterative logic easier to reason about.

        Time : O(n)  — full BFS + two linear ancestor walks
        Space: O(n)  — parent map + ancestor set
        """
        if root is None:
            return None

        parent: dict[TreeNode, TreeNode | None] = {root: None}
        queue: deque[TreeNode] = deque([root])

        while queue:
            node = queue.popleft()
            if node.left:
                parent[node.left] = node
                queue.append(node.left)
            if node.right:
                parent[node.right] = node
                queue.append(node.right)

        p_ancestors: set[TreeNode] = set()
        current: TreeNode | None = p
        while current is not None:
            p_ancestors.add(current)
            current = parent[current]

        current = q
        while current is not None:
            if current in p_ancestors:
                return current
            current = parent[current]

        return current

    @staticmethod
    def find_safe(
        root: TreeNode | None,
        p: TreeNode,
        q: TreeNode,
    ) -> TreeNode:
        """
        LCA variant for when p and/or q might not exist in the tree.

        The standard find() is ambiguous when a target is absent:
            - It returns non-None if it finds ONE target (the other absent),
              which looks identical to "one target found, other is its descendant."

        Fix: track how many targets have been confirmed found.
        Only return the LCA if both were found (found_count == 2).

        Implementation: use a mutable container (list) as a side channel
        to communicate found_count out of the inner recursive function
        without changing its return-value semantics.

        Time : O(n)
        Space: O(h)
        """
        found_count = [0]  # mutable so it can be modified in the inner closure.

        def recurse(node: TreeNode | None) -> TreeNode | None:
            if node is None:
                return None

            left_result = recurse(node=node.left)
            right_result = recurse(node=node.right)

            if node is p:
                found_count[0] += 1
            if node is q:
                found_count[0] += 1
            if node is p or node is q:
                return node

            if left_result is not None and right_result is not None:
                return node

            return left_result if left_result is not None else right_result

        candidate = recurse(node=root)
        return candidate if found_count[0] == 2 else None


if __name__ == "__main__":
    A = TreeNode(1)
    B = TreeNode(2)
    C = TreeNode(3)
    D = TreeNode(4)
    E = TreeNode(5)
    F = TreeNode(6)
    G = TreeNode(7)
    H = TreeNode(8)

    A.left = B
    A.right = C
    B.left = D
    B.right = E
    D.left = G
    D.right = H
    A.right = C
    C.right = F

    # This is what the above tree would look like:
    #        1
    #       / \
    #      2  3
    #     / \  \
    #    4  5  6
    #   / \
    #  7  8

    lca = TreeLCA()
    result = lca.find(root=A, p=H, q=E)
    print(result)
    result = lca.find_iterative(root=A, p=H, q=E)
    print(result)

    print("====== Safe Find ======")
    result = lca.find_safe(root=A, p=H, q=E)
    print(result)

    Z = TreeNode(val=10)
    result = lca.find_safe(root=A, p=H, q=Z)
    print(result)
````

## Binary Search Tree

A binary tree is called a binary search tree if the value at any node in the tree is
greater than the values in all the nodes of its left subtree and less than or equal to
the values of f all the nodes of the right subtree.

### The Invariant

A BST is a binary tree where, for every node `n`:

- every value in `n.left`'s subtree is `< n.value`
- every value in `n.rights`'s subtree is `> (>=) n.value`

The difference between bst invariant and heap invariant, is that a heap only constraints
a node against its children (local, one hop), but bst invariant is global, it
constraints a node against every node in both subtrees. This difference buys us
something that a heap can't give us: an ordered traversal. In-order traversal of a BST
yields values in sorted order, for free, in O(n).

**A key insight to binary search trees, is understanding that a binary search tree is
actually a binary search, restructured as a data structure, to make insertion and
deletion efficient too.**

| Structure | Search | Insert | Delete |
|---|---|---|---|
| Unsorted array | O(n) | O(1) | O(n) |
| Sorted array | O(log n) | O(n) (shifting) | O(n) (shifting) |
| BST (balanced) | O(h) = O(log n) | O(h) = O(log n) | O(h) = O(log n) |
| BST (degenerate) | O(h) = O(n) | O(n) | O(n) |
