# Algorithms Techniques

## Two Pointers Technique

The two pointers technique is a powerful strategy when two indices(pointers) traverse a
data structure either toward each other or in the same direction to solve problems more
efficiently. They usually can reduce the O(n^2) time complexity to O(n).

As the title says, the technique is the use of two pointers(mostly tracking sequence 
indexes) to solve a problem. You can think of a pointer, as a reference to an object.

There are two common patterns while using the technique:

- fast and slow pointers: One pointer moves at slow pace, while other one moves faster.
- start and end pointers: One starting from the start and the other one starts from the
  end and they keep traversing until they reach each other.

### When We Should you Use it?

When observing or tracking two elements of a the collection, will save another
traversing or using another data structure.

### Use Cases

Ok, at this point you are reviewing this(and even when I am writing this:)) I have
solved tones of problems with this technique, but here are two use cases:

- Finding a cycle in a linked list: Move the fast pointer twice the speed of the slow
  pointer and voila! You'll find the cycle.
- Finding two numbers which will sum up to a target number in a sorted array: Move both
  start and end pointers until the numbers are found or they reach each other.
